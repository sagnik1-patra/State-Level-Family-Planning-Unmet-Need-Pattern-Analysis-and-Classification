```python
import os
import json
import warnings
import pickle

import numpy as np
import pandas as pd
import yaml

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =====================================================================
# 2. PROJECT PATHS
# =====================================================================

PROJECT_DIR = r"C:\Users\sagni\Downloads\State-Level Family Planning Unmet Need Pattern Analysis and Classification"

DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "unmetneedforFPNFHS3.csv"
)

H5_PATH = os.path.join(
    PROJECT_DIR,
    "family_planning_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "family_planning_preprocessor.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "family_planning_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "family_planning_metadata.json"
)

os.makedirs(PROJECT_DIR, exist_ok=True)


# =====================================================================
# 3. LOAD DATASET
# =====================================================================

print("=" * 75)
print("STATE-LEVEL FAMILY PLANNING UNMET NEED ANALYSIS")
print("=" * 75)

print("\nLoading dataset...")

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)


# =====================================================================
# 4. DISPLAY BASIC DATASET INFORMATION
# =====================================================================

print("\n" + "=" * 75)
print("DATASET INFORMATION")
print("=" * 75)

print("\nColumns:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")

print("\nFirst 5 rows:")
print(df.head())

print("\nDataset data types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())


# =====================================================================
# 5. CLEAN COLUMN NAMES
# =====================================================================

original_columns = df.columns.tolist()

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

print("\nCleaned column names:")

for col in df.columns:
    print(" -", col)


# =====================================================================
# 6. REMOVE COMPLETELY EMPTY COLUMNS
# =====================================================================

df = df.dropna(axis=1, how="all")

print("\nShape after removing empty columns:")
print(df.shape)


# =====================================================================
# 7. IDENTIFY STATE / REGION COLUMN
# =====================================================================

state_column = None

possible_state_names = [
    "state",
    "states",
    "state/ut",
    "state / ut",
    "states/uts",
    "states/uts",
    "india/states/uts",
    "india / states / uts",
    "region",
    "area"
]

for col in df.columns:

    col_lower = col.lower().strip()

    if col_lower in possible_state_names:
        state_column = col
        break

    if "state" in col_lower or "region" in col_lower:
        state_column = col
        break


# If not automatically detected, find first text column

if state_column is None:

    object_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    if len(object_columns) > 0:
        state_column = object_columns[0]


print("\nDetected State/Region column:")
print(state_column)


# =====================================================================
# 8. CONVERT POSSIBLE NUMERIC COLUMNS
# =====================================================================

print("\nConverting columns to numeric where possible...")

for col in df.columns:

    if col == state_column:
        continue

    if df[col].dtype == "object":

        cleaned = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
            .str.replace("%", "", regex=False)
        )

        numeric_version = pd.to_numeric(
            cleaned,
            errors="coerce"
        )

        # Convert to numeric only if majority of non-null values
        # can be interpreted as numbers.

        original_non_null = df[col].notna().sum()
        numeric_non_null = numeric_version.notna().sum()

        if (
            original_non_null > 0
            and numeric_non_null / original_non_null >= 0.60
        ):
            df[col] = numeric_version


# =====================================================================
# 9. GET NUMERIC COLUMNS
# =====================================================================

numeric_columns = df.select_dtypes(
    include=[np.number]
).columns.tolist()

print("\nNumeric columns detected:")

for col in numeric_columns:
    print(" -", col)

if len(numeric_columns) == 0:
    raise ValueError(
        "No numeric columns were detected in the dataset."
    )


# =====================================================================
# 10. IDENTIFY TOTAL UNMET NEED COLUMN
# =====================================================================

target_source_column = None


# Priority 1:
# Column containing "unmet" and "total"

for col in numeric_columns:

    lower = col.lower()

    if "unmet" in lower and "total" in lower:
        target_source_column = col
        break


# Priority 2:
# Any column containing "total"

if target_source_column is None:

    for col in numeric_columns:

        if "total" in col.lower():
            target_source_column = col
            break


# Priority 3:
# First numeric column

if target_source_column is None:
    target_source_column = numeric_columns[0]


print("\nTarget source column:")
print(target_source_column)


# =====================================================================
# 11. CLEAN MISSING TARGET VALUES
# =====================================================================

df = df.dropna(
    subset=[target_source_column]
).reset_index(drop=True)

print("\nRows after target cleaning:")
print(len(df))


# =====================================================================
# 12. CREATE CLASSIFICATION TARGET
# =====================================================================
#
# Target:
#
# 0 = Low
# 1 = Moderate
# 2 = High
#
# Quantile-based classification is used because it automatically adapts
# to the distribution of the dataset.
# =====================================================================

print("\nCreating classification target...")

target_values = df[target_source_column].astype(float)

try:

    df["Unmet_Need_Category"] = pd.qcut(
        target_values,
        q=3,
        labels=[
            "Low",
            "Moderate",
            "High"
        ],
        duplicates="drop"
    )

except Exception:

    q1 = target_values.quantile(0.33)
    q2 = target_values.quantile(0.66)

    def categorize(value):

        if value <= q1:
            return "Low"

        elif value <= q2:
            return "Moderate"

        else:
            return "High"

    df["Unmet_Need_Category"] = (
        target_values.apply(categorize)
    )


print("\nTarget distribution:")
print(
    df["Unmet_Need_Category"]
    .value_counts()
)


# =====================================================================
# 13. LABEL ENCODER
# =====================================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df["Unmet_Need_Category"].astype(str)
)

class_names = label_encoder.classes_.tolist()

print("\nClasses:")
print(class_names)

print("\nClass mapping:")

for index, name in enumerate(class_names):
    print(f"{index} -> {name}")


# =====================================================================
# 14. FEATURE ENGINEERING
# =====================================================================

print("\n" + "=" * 75)
print("FEATURE ENGINEERING")
print("=" * 75)


# ---------------------------------------------------------------
# Use numeric variables except target source.
#
# IMPORTANT:
# We remove target_source_column because using the same value
# from which the classification target was created would cause
# target leakage.
# ---------------------------------------------------------------

feature_columns = [
    col
    for col in numeric_columns
    if col != target_source_column
]


if len(feature_columns) == 0:
    raise ValueError(
        "No predictor features remain after removing target column."
    )


X = df[feature_columns].copy()


print("\nFeatures used:")

for col in feature_columns:
    print(" -", col)


print("\nNumber of input features:")
print(len(feature_columns))


# =====================================================================
# 15. HANDLE INFINITE VALUES
# =====================================================================

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


# =====================================================================
# 16. MISSING VALUE IMPUTATION
# =====================================================================

imputer = SimpleImputer(
    strategy="median"
)

X_imputed = imputer.fit_transform(X)


# =====================================================================
# 17. FEATURE SCALING
# =====================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X_imputed
)


print("\nFeature preprocessing completed.")
print("Processed feature shape:", X_scaled.shape)


# =====================================================================
# 18. TRAIN TEST SPLIT
# =====================================================================

print("\n" + "=" * 75)
print("TRAIN TEST SPLIT")
print("=" * 75)


# For very small datasets stratification can occasionally fail.
# Therefore, check the minimum class size first.

unique_classes, class_counts = np.unique(
    y,
    return_counts=True
)

minimum_class_count = class_counts.min()


if minimum_class_count >= 2:

    stratify_option = y

else:

    stratify_option = None


X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.25,
    random_state=42,
    stratify=stratify_option
)


print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# =====================================================================
# 19. BUILD NEURAL NETWORK
# =====================================================================

print("\n" + "=" * 75)
print("BUILDING NEURAL NETWORK")
print("=" * 75)

number_of_features = X_train.shape[1]
number_of_classes = len(class_names)


model = Sequential([
    
    tf.keras.Input(
        shape=(number_of_features,)
    ),

    Dense(
        64,
        activation="relu"
    ),

    BatchNormalization(),

    Dropout(0.25),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )
])


# =====================================================================
# 20. COMPILE MODEL
# =====================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


print("\nModel summary:")

model.summary()


# =====================================================================
# 21. EARLY STOPPING
# =====================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True,

    verbose=1
)


# =====================================================================
# 22. TRAIN MODEL
# =====================================================================

print("\n" + "=" * 75)
print("TRAINING MODEL")
print("=" * 75)


history = model.fit(

    X_train,
    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=4,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# =====================================================================
# 23. MODEL EVALUATION
# =====================================================================

print("\n" + "=" * 75)
print("MODEL EVALUATION")
print("=" * 75)


test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)


print(
    f"\nTest Loss     : {test_loss:.4f}"
)

print(
    f"Test Accuracy : {test_accuracy:.4f}"
)

print(
    f"Test Accuracy : {test_accuracy * 100:.2f}%"
)


# =====================================================================
# 24. GENERATE PREDICTIONS
# =====================================================================

prediction_probabilities = model.predict(
    X_test,
    verbose=0
)

y_pred = np.argmax(
    prediction_probabilities,
    axis=1
)


# =====================================================================
# 25. CLASSIFICATION REPORT
# =====================================================================

print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        labels=np.arange(number_of_classes),
        target_names=class_names,
        zero_division=0
    )
)


# =====================================================================
# 26. CONFUSION MATRIX
# =====================================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=np.arange(number_of_classes)
)

print("\nConfusion Matrix:")

print(cm)


# =====================================================================
# 27. ACCURACY USING SKLEARN
# =====================================================================

sklearn_accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    "\nSklearn Accuracy:",
    round(sklearn_accuracy, 4)
)


# =====================================================================
# 28. SAVE H5 MODEL
# =====================================================================

print("\n" + "=" * 75)
print("SAVING PROJECT FILES")
print("=" * 75)


model.save(
    H5_PATH
)

print(
    "\nSaved H5 model:"
)

print(H5_PATH)


# =====================================================================
# 29. SAVE PKL PREPROCESSOR
# =====================================================================
#
# The PKL contains everything needed to prepare future input:
#
# - feature names
# - imputer
# - scaler
# - label encoder
# - target information
#
# =====================================================================

preprocessor = {

    "feature_columns":
        feature_columns,

    "imputer":
        imputer,

    "scaler":
        scaler,

    "label_encoder":
        label_encoder,

    "class_names":
        class_names,

    "target_source_column":
        target_source_column,

    "state_column":
        state_column
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        preprocessor,
        file
    )


print(
    "\nSaved PKL preprocessor:"
)

print(PKL_PATH)


# =====================================================================
# 30. SAVE YAML CONFIGURATION
# =====================================================================

config = {

    "project": {
        "name":
            "State-Level Family Planning Unmet Need Pattern Analysis and Classification",

        "dataset":
            DATASET_PATH,

        "project_directory":
            PROJECT_DIR
    },

    "data": {

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1]),

        "state_column":
            str(state_column),

        "target_source":
            str(target_source_column),

        "classification_target":
            "Unmet_Need_Category",

        "features":
            [str(x) for x in feature_columns]
    },

    "model": {

        "type":
            "Artificial Neural Network",

        "framework":
            "TensorFlow/Keras",

        "hidden_layers":
            [64, 32, 16],

        "activation":
            "ReLU",

        "output_activation":
            "Softmax",

        "optimizer":
            "Adam",

        "loss":
            "Sparse Categorical Crossentropy",

        "epochs":
            200,

        "batch_size":
            4
    },

    "classification": {

        "method":
            "Quantile-based classification",

        "categories":
            [str(x) for x in class_names]
    },

    "evaluation": {

        "test_accuracy":
            float(test_accuracy),

        "test_loss":
            float(test_loss)
    },

    "output_files": {

        "h5_model":
            H5_PATH,

        "pkl_preprocessor":
            PKL_PATH,

        "yaml_config":
            YAML_PATH,

        "json_metadata":
            JSON_PATH
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nSaved YAML configuration:"
)

print(YAML_PATH)


# =====================================================================
# 31. SAVE JSON METADATA
# =====================================================================

classification_dictionary = (
    classification_report(
        y_test,
        y_pred,
        labels=np.arange(number_of_classes),
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )
)


metadata = {

    "project_name":
        "State-Level Family Planning Unmet Need Pattern Analysis and Classification",

    "dataset_path":
        DATASET_PATH,

    "dataset_rows":
        int(df.shape[0]),

    "dataset_columns":
        int(df.shape[1]),

    "original_columns":
        [str(x) for x in original_columns],

    "processed_columns":
        [str(x) for x in df.columns.tolist()],

    "state_column":
        str(state_column),

    "target_source_column":
        str(target_source_column),

    "target_column":
        "Unmet_Need_Category",

    "feature_columns":
        [str(x) for x in feature_columns],

    "number_of_features":
        int(len(feature_columns)),

    "classes":
        [str(x) for x in class_names],

    "training_samples":
        int(len(X_train)),

    "testing_samples":
        int(len(X_test)),

    "test_accuracy":
        float(test_accuracy),

    "test_loss":
        float(test_loss),

    "classification_report":
        classification_dictionary,

    "confusion_matrix":
        cm.tolist(),

    "model_file":
        H5_PATH,

    "preprocessor_file":
        PKL_PATH,

    "configuration_file":
        YAML_PATH
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "\nSaved JSON metadata:"
)

print(JSON_PATH)


# =====================================================================
# 32. VERIFY SAVED PKL
# =====================================================================

print("\n" + "=" * 75)
print("VERIFYING PKL FILE")
print("=" * 75)


with open(
    PKL_PATH,
    "rb"
) as file:

    loaded_preprocessor = pickle.load(file)


print(
    "\nPKL loaded successfully."
)

print(
    "Saved features:",
    loaded_preprocessor[
        "feature_columns"
    ]
)


# =====================================================================
# 33. VERIFY SAVED H5 MODEL
# =====================================================================

print("\n" + "=" * 75)
print("VERIFYING H5 MODEL")
print("=" * 75)


loaded_model = tf.keras.models.load_model(
    H5_PATH
)


verification_probabilities = (
    loaded_model.predict(
        X_test[:1],
        verbose=0
    )
)


verification_prediction = np.argmax(
    verification_probabilities,
    axis=1
)[0]


verification_label = (
    label_encoder.inverse_transform(
        [verification_prediction]
    )[0]
)


print(
    "\nH5 model loaded successfully."
)

print(
    "Sample predicted class:",
    verification_label
)


# =====================================================================
# 34. DISPLAY OUTPUT FILES
# =====================================================================

print("\n" + "=" * 75)
print("GENERATED FILES")
print("=" * 75)


generated_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]


for path in generated_files:

    if os.path.exists(path):

        size_kb = (
            os.path.getsize(path)
            / 1024
        )

        print(
            f"\n[OK] {os.path.basename(path)}"
        )

        print(
            f"     Size: {size_kb:.2f} KB"
        )

    else:

        print(
            f"\n[ERROR] {os.path.basename(path)}"
        )


# =====================================================================
# 35. FINAL SUMMARY
# =====================================================================

print("\n" + "=" * 75)
print("PROJECT SUMMARY")
print("=" * 75)


print(
    f"""
Project:
State-Level Family Planning Unmet Need Pattern Analysis and Classification

Dataset:
{DATASET_PATH}

Target source:
{target_source_column}

Classification target:
Unmet_Need_Category

Classes:
{class_names}

Features:
{len(feature_columns)}

Training samples:
{len(X_train)}

Testing samples:
{len(X_test)}

Test Accuracy:
{test_accuracy * 100:.2f}%

Generated files:

1. family_planning_model.h5
2. family_planning_preprocessor.pkl
3. family_planning_config.yaml
4. family_planning_metadata.json

Output directory:
{PROJECT_DIR}
"""
)


print("=" * 75)
print("EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 75)
```
