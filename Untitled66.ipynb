```python
import os
import json
import pickle
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yaml

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.callbacks import (
    EarlyStopping
)

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.impute import (
    SimpleImputer
)

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import (
    DecisionTreeClassifier
)

from sklearn.neighbors import (
    KNeighborsClassifier
)

from sklearn.linear_model import (
    LogisticRegression
)

from sklearn.svm import (
    SVC
)


warnings.filterwarnings(
    "ignore"
)


# ======================================================================
# 2. RANDOM SEEDS
# ======================================================================

RANDOM_STATE = 42

np.random.seed(
    RANDOM_STATE
)

random.seed(
    RANDOM_STATE
)

tf.random.set_seed(
    RANDOM_STATE
)


# ======================================================================
# 3. PROJECT DIRECTORY
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\State-Level Family Planning Unmet Need Pattern Analysis and Classification"
)


DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "unmetneedforFPNFHS3.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# 4. OUTPUT PATHS
# ======================================================================

H5_PATH = os.path.join(
    PROJECT_DIR,
    "pso_family_planning_model.h5"
)


PKL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_family_planning_preprocessor.pkl"
)


YAML_PATH = os.path.join(
    PROJECT_DIR,
    "pso_family_planning_config.yaml"
)


JSON_PATH = os.path.join(
    PROJECT_DIR,
    "pso_family_planning_metadata.json"
)


ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_accuracy_graph.png"
)


HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "pso_heatmap.png"
)


COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_comparison_graph.png"
)


RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result.csv"
)


RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result_graph.png"
)


PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction.csv"
)


PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction_graph.png"
)


FEATURE_SELECTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection.csv"
)


FEATURE_SELECTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection_graph.png"
)


FITNESS_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_fitness_graph.png"
)


# ======================================================================
# 5. LOAD DATASET
# ======================================================================

print(
    "=" * 80
)

print(
    "PSO-BASED FAMILY PLANNING UNMET NEED ANALYSIS"
)

print(
    "=" * 80
)


if not os.path.exists(
    DATASET_PATH
):

    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}"
    )


df = pd.read_csv(
    DATASET_PATH
)


print(
    "\nDataset loaded successfully."
)

print(
    "Original dataset shape:",
    df.shape
)


print(
    "\nOriginal columns:"
)


for index, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{index}. {column}"
    )


# ======================================================================
# 6. SAVE ORIGINAL COLUMN NAMES
# ======================================================================

original_columns = (
    df.columns
    .astype(str)
    .tolist()
)


# ======================================================================
# 7. CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        "\n",
        " ",
        regex=False
    )
    .str.replace(
        "\r",
        " ",
        regex=False
    )
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


# Remove completely empty columns

df = df.dropna(
    axis=1,
    how="all"
)


print(
    "\nDataset shape after cleaning:",
    df.shape
)


# ======================================================================
# 8. DETECT STATE / REGION COLUMN
# ======================================================================

state_column = None


for column in df.columns:

    column_lower = (
        column
        .lower()
        .strip()
    )


    if (
        "state" in column_lower
        or
        "region" in column_lower
        or
        "area" in column_lower
    ):

        state_column = column

        break


# If automatic detection fails,
# use first text column.

if state_column is None:

    object_columns = (
        df.select_dtypes(
            include=[
                "object",
                "string"
            ]
        )
        .columns
        .tolist()
    )


    if len(
        object_columns
    ) > 0:

        state_column = (
            object_columns[0]
        )


print(
    "\nDetected State/Region column:"
)

print(
    state_column
)


# ======================================================================
# 9. CONVERT POSSIBLE NUMERIC COLUMNS
# ======================================================================

print(
    "\nConverting possible numeric columns..."
)


for column in df.columns:


    if column == state_column:

        continue


    if df[column].dtype == "object":


        cleaned_column = (

            df[column]
            .astype(str)
            .str.strip()
            .str.replace(
                ",",
                "",
                regex=False
            )
            .str.replace(
                "%",
                "",
                regex=False
            )
        )


        numeric_version = (
            pd.to_numeric(
                cleaned_column,
                errors="coerce"
            )
        )


        original_non_null = (
            df[column]
            .notna()
            .sum()
        )


        numeric_non_null = (
            numeric_version
            .notna()
            .sum()
        )


        if (
            original_non_null > 0
            and
            numeric_non_null
            / original_non_null
            >= 0.60
        ):

            df[column] = (
                numeric_version
            )


# ======================================================================
# 10. FIND NUMERIC COLUMNS
# ======================================================================

numeric_columns = (
    df.select_dtypes(
        include=[
            np.number
        ]
    )
    .columns
    .tolist()
)


print(
    "\nNumeric columns:"
)


for column in numeric_columns:

    print(
        " -",
        column
    )


if len(
    numeric_columns
) < 2:

    raise ValueError(
        "At least two numeric columns are required."
    )


# ======================================================================
# 11. DETECT TARGET SOURCE COLUMN
# ======================================================================

target_source_column = None


# First preference:
# column containing "unmet" AND "total"

for column in numeric_columns:


    lower = (
        column.lower()
    )


    if (
        "unmet" in lower
        and
        "total" in lower
    ):

        target_source_column = (
            column
        )

        break


# Second preference:
# any column containing "total"

if target_source_column is None:


    for column in numeric_columns:


        if (
            "total"
            in column.lower()
        ):

            target_source_column = (
                column
            )

            break


# Fallback:
# first numeric column

if target_source_column is None:

    target_source_column = (
        numeric_columns[0]
    )


print(
    "\nTarget source column:"
)

print(
    target_source_column
)


# ======================================================================
# 12. REMOVE ROWS WITH MISSING TARGET
# ======================================================================

df = (

    df.dropna(
        subset=[
            target_source_column
        ]
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nRows after target cleaning:",
    len(df)
)


# ======================================================================
# 13. CREATE CLASSIFICATION TARGET
# ======================================================================

target_values = (

    df[
        target_source_column
    ]

    .astype(float)
)


lower_threshold = (

    target_values
    .quantile(
        1 / 3
    )
)


upper_threshold = (

    target_values
    .quantile(
        2 / 3
    )
)


def create_category(
    value
):


    if value <= lower_threshold:

        return "Low"


    elif value <= upper_threshold:

        return "Moderate"


    else:

        return "High"


df[
    "Unmet_Need_Category"
] = (

    target_values
    .apply(
        create_category
    )
)


print(
    "\nClassification thresholds:"
)


print(
    f"Low <= "
    f"{lower_threshold:.3f}"
)


print(
    f"Moderate > "
    f"{lower_threshold:.3f} "
    f"and <= "
    f"{upper_threshold:.3f}"
)


print(
    f"High > "
    f"{upper_threshold:.3f}"
)


print(
    "\nTarget distribution:"
)


print(

    df[
        "Unmet_Need_Category"
    ]

    .value_counts()
)


# ======================================================================
# 14. LABEL ENCODE TARGET
# ======================================================================

label_encoder = (
    LabelEncoder()
)


y = (

    label_encoder
    .fit_transform(

        df[
            "Unmet_Need_Category"
        ]
        .astype(str)
    )
)


class_names = (

    label_encoder
    .classes_
    .tolist()
)


print(
    "\nClass mapping:"
)


for index, class_name in enumerate(
    class_names
):

    print(
        index,
        "=",
        class_name
    )


# ======================================================================
# 15. PREPARE CANDIDATE FEATURES
# ======================================================================

# IMPORTANT:
#
# The target source column is excluded because the target categories
# were generated directly from it.
#
# Including it would cause target leakage.

feature_columns = [

    column

    for column in numeric_columns

    if column
    !=
    target_source_column
]


if len(
    feature_columns
) == 0:

    raise ValueError(
        "No predictor features remain."
    )


print(
    "\nCandidate features for PSO:"
)


for feature in feature_columns:

    print(
        " -",
        feature
    )


NUMBER_OF_FEATURES = (
    len(
        feature_columns
    )
)


# ======================================================================
# 16. CREATE FEATURE DATAFRAME
# ======================================================================

X_dataframe = (

    df[
        feature_columns
    ]

    .copy()
)


X_dataframe = (

    X_dataframe
    .replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )
)


# ======================================================================
# 17. IMPUTE MISSING VALUES
# ======================================================================

initial_imputer = (

    SimpleImputer(
        strategy="median"
    )
)


X_imputed = (

    initial_imputer
    .fit_transform(
        X_dataframe
    )
)


# ======================================================================
# 18. SCALE FEATURES
# ======================================================================

initial_scaler = (

    StandardScaler()
)


X_all_scaled = (

    initial_scaler
    .fit_transform(
        X_imputed
    )
)


print(
    "\nInitial feature matrix shape:",
    X_all_scaled.shape
)


# ======================================================================
# 19. PSO CONFIGURATION
# ======================================================================

print(
    "\n"
    + "=" * 80
)

print(
    "PARTICLE SWARM OPTIMIZATION FEATURE SELECTION"
)

print(
    "=" * 80
)


# Number of particles in the swarm

SWARM_SIZE = 30


# Number of optimization iterations

ITERATIONS = 30


# Inertia weight

INERTIA_WEIGHT = 0.729


# Cognitive coefficient
# Attraction toward personal best

COGNITIVE_COEFFICIENT = 1.49445


# Social coefficient
# Attraction toward global best

SOCIAL_COEFFICIENT = 1.49445


# Maximum absolute velocity

MAX_VELOCITY = 6.0


# Feature penalty
#
# This discourages PSO from simply selecting every feature.

FEATURE_PENALTY = 0.02


print(
    "\nPSO Configuration:"
)


print(
    "Swarm Size:",
    SWARM_SIZE
)


print(
    "Iterations:",
    ITERATIONS
)


print(
    "Inertia Weight:",
    INERTIA_WEIGHT
)


print(
    "Cognitive Coefficient:",
    COGNITIVE_COEFFICIENT
)


print(
    "Social Coefficient:",
    SOCIAL_COEFFICIENT
)


print(
    "Feature Penalty:",
    FEATURE_PENALTY
)


print(
    "Candidate Features:",
    NUMBER_OF_FEATURES
)


# ======================================================================
# 20. DETERMINE CROSS-VALIDATION FOLDS
# ======================================================================

class_counts = (
    np.bincount(
        y
    )
)


minimum_class_count = (
    class_counts.min()
)


if minimum_class_count >= 3:

    CV_SPLITS = 3


elif minimum_class_count >= 2:

    CV_SPLITS = 2


else:

    CV_SPLITS = None


print(
    "\nPSO Cross Validation Folds:",
    CV_SPLITS
)


# ======================================================================
# 21. SIGMOID FUNCTION
# ======================================================================

def sigmoid(
    value
):

    value = np.clip(
        value,
        -50,
        50
    )


    return (
        1.0
        /
        (
            1.0
            +
            np.exp(
                -value
            )
        )
    )


# ======================================================================
# 22. ENSURE VALID FEATURE MASK
# ======================================================================

def ensure_valid_position(
    position
):


    position = (
        position
        .astype(int)
        .copy()
    )


    # PSO cannot evaluate an empty feature subset.

    if (
        position.sum()
        ==
        0
    ):


        random_index = (
            np.random.randint(
                0,
                len(
                    position
                )
            )
        )


        position[
            random_index
        ] = 1


    return position


# ======================================================================
# 23. FITNESS CACHE
# ======================================================================

fitness_cache = {}


# ======================================================================
# 24. PSO FITNESS FUNCTION
# ======================================================================

def calculate_fitness(
    particle_position
):


    particle_position = (
        ensure_valid_position(
            particle_position
        )
    )


    key = tuple(
        particle_position
        .tolist()
    )


    # Reuse previously calculated fitness.

    if key in fitness_cache:

        return (
            fitness_cache[
                key
            ]
        )


    selected_indices = (

        np.where(
            particle_position
            ==
            1
        )[0]
    )


    if len(
        selected_indices
    ) == 0:

        return 0.0


    X_selected = (

        X_all_scaled[
            :,
            selected_indices
        ]
    )


    # Random Forest is used as the fitness evaluator.

    evaluator = (

        RandomForestClassifier(

            n_estimators=100,

            random_state=RANDOM_STATE,

            class_weight="balanced"
        )
    )


    try:


        # ==============================================================
        # CROSS VALIDATION
        # ==============================================================

        if CV_SPLITS is not None:


            cross_validator = (

                StratifiedKFold(

                    n_splits=CV_SPLITS,

                    shuffle=True,

                    random_state=RANDOM_STATE
                )
            )


            scores = (

                cross_val_score(

                    evaluator,

                    X_selected,

                    y,

                    cv=cross_validator,

                    scoring="accuracy"
                )
            )


            mean_accuracy = (
                scores.mean()
            )


        # ==============================================================
        # FALLBACK HOLDOUT
        # ==============================================================

        else:


            (
                X_train_temp,
                X_test_temp,
                y_train_temp,
                y_test_temp
            ) = train_test_split(

                X_selected,
                y,

                test_size=0.25,

                random_state=RANDOM_STATE
            )


            evaluator.fit(
                X_train_temp,
                y_train_temp
            )


            mean_accuracy = (

                evaluator.score(
                    X_test_temp,
                    y_test_temp
                )
            )


        # ==============================================================
        # FEATURE REDUCTION PENALTY
        # ==============================================================

        selected_feature_ratio = (

            len(
                selected_indices
            )
            /
            NUMBER_OF_FEATURES
        )


        fitness = (

            mean_accuracy

            -

            FEATURE_PENALTY
            *
            selected_feature_ratio
        )


    except Exception:


        fitness = 0.0


    fitness_cache[
        key
    ] = float(
        fitness
    )


    return float(
        fitness
    )


# ======================================================================
# 25. INITIALIZE PARTICLES
# ======================================================================

print(
    "\nInitializing PSO swarm..."
)


# Binary particle positions

positions = (

    np.random.randint(

        0,
        2,

        size=(
            SWARM_SIZE,
            NUMBER_OF_FEATURES
        )
    )
)


# Ensure no particle contains zero features.

for particle_index in range(
    SWARM_SIZE
):

    positions[
        particle_index
    ] = (

        ensure_valid_position(

            positions[
                particle_index
            ]
        )
    )


# Continuous particle velocities

velocities = (

    np.random.uniform(

        low=-1.0,

        high=1.0,

        size=(
            SWARM_SIZE,
            NUMBER_OF_FEATURES
        )
    )
)


# ======================================================================
# 26. INITIAL PERSONAL BESTS
# ======================================================================

personal_best_positions = (
    positions.copy()
)


personal_best_fitness = (

    np.full(
        SWARM_SIZE,
        -np.inf
    )
)


# ======================================================================
# 27. INITIAL GLOBAL BEST
# ======================================================================

global_best_position = None

global_best_fitness = -np.inf


# ======================================================================
# 28. PSO HISTORY
# ======================================================================

global_best_history = []

average_fitness_history = []

selected_feature_history = []


# ======================================================================
# 29. RUN BINARY PSO
# ======================================================================

print(
    "\nStarting Binary PSO..."
)


for iteration in range(
    ITERATIONS
):


    current_fitness_values = []


    # ==================================================================
    # EVALUATE PARTICLES
    # ==================================================================

    for particle_index in range(
        SWARM_SIZE
    ):


        current_position = (

            positions[
                particle_index
            ]
        )


        fitness = (

            calculate_fitness(
                current_position
            )
        )


        current_fitness_values.append(
            fitness
        )


        # ==============================================================
        # UPDATE PERSONAL BEST
        # ==============================================================

        if (
            fitness
            >
            personal_best_fitness[
                particle_index
            ]
        ):


            personal_best_fitness[
                particle_index
            ] = fitness


            personal_best_positions[
                particle_index
            ] = (

                current_position.copy()
            )


        # ==============================================================
        # UPDATE GLOBAL BEST
        # ==============================================================

        if (
            fitness
            >
            global_best_fitness
        ):


            global_best_fitness = (
                fitness
            )


            global_best_position = (

                current_position.copy()
            )


    # ==================================================================
    # ITERATION STATISTICS
    # ==================================================================

    average_fitness = (

        np.mean(
            current_fitness_values
        )
    )


    global_best_history.append(
        global_best_fitness
    )


    average_fitness_history.append(
        average_fitness
    )


    selected_feature_history.append(

        int(
            global_best_position.sum()
        )
    )


    print(

        f"Iteration "
        f"{iteration + 1:02d}/"
        f"{ITERATIONS} | "

        f"Global Best Fitness: "
        f"{global_best_fitness:.4f} | "

        f"Average Fitness: "
        f"{average_fitness:.4f} | "

        f"Selected Features: "
        f"{global_best_position.sum()}"
    )


    # ==================================================================
    # UPDATE PARTICLE VELOCITIES AND POSITIONS
    # ==================================================================

    for particle_index in range(
        SWARM_SIZE
    ):


        r1 = (

            np.random.random(
                NUMBER_OF_FEATURES
            )
        )


        r2 = (

            np.random.random(
                NUMBER_OF_FEATURES
            )
        )


        # ==============================================================
        # COGNITIVE COMPONENT
        # ==============================================================

        cognitive_component = (

            COGNITIVE_COEFFICIENT

            *

            r1

            *

            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )
        )


        # ==============================================================
        # SOCIAL COMPONENT
        # ==============================================================

        social_component = (

            SOCIAL_COEFFICIENT

            *

            r2

            *

            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )
        )


        # ==============================================================
        # VELOCITY UPDATE
        # ==============================================================

        velocities[
            particle_index
        ] = (

            INERTIA_WEIGHT
            *
            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )


        # Restrict velocity.

        velocities[
            particle_index
        ] = (

            np.clip(

                velocities[
                    particle_index
                ],

                -MAX_VELOCITY,

                MAX_VELOCITY
            )
        )


        # ==============================================================
        # BINARY POSITION UPDATE
        # ==============================================================

        probabilities = (

            sigmoid(

                velocities[
                    particle_index
                ]
            )
        )


        random_values = (

            np.random.random(
                NUMBER_OF_FEATURES
            )
        )


        new_position = (

            random_values
            <
            probabilities
        ).astype(int)


        new_position = (

            ensure_valid_position(
                new_position
            )
        )


        positions[
            particle_index
        ] = (
            new_position
        )


# ======================================================================
# 30. FINAL PSO FEATURE SUBSET
# ======================================================================

best_particle = (

    global_best_position
    .copy()
)


selected_indices = (

    np.where(
        best_particle
        ==
        1
    )[0]
)


selected_features = [

    feature_columns[
        index
    ]

    for index in selected_indices
]


rejected_features = [

    feature_columns[
        index
    ]

    for index in range(
        NUMBER_OF_FEATURES
    )

    if best_particle[
        index
    ] == 0
]


print(
    "\n"
    + "=" * 80
)

print(
    "PSO FEATURE SELECTION RESULT"
)

print(
    "=" * 80
)


print(
    "\nBest PSO Fitness:",
    round(
        global_best_fitness,
        4
    )
)


print(
    "Total Candidate Features:",
    NUMBER_OF_FEATURES
)


print(
    "Selected Features:",
    len(
        selected_features
    )
)


print(
    "Rejected Features:",
    len(
        rejected_features
    )
)


print(
    "\nPSO Selected Features:"
)


for feature in selected_features:

    print(
        " [SELECTED]",
        feature
    )


print(
    "\nPSO Rejected Features:"
)


if len(
    rejected_features
) == 0:

    print(
        " None"
    )


else:

    for feature in rejected_features:

        print(
            " [REJECTED]",
            feature
        )


# ======================================================================
# 31. SAVE PSO FEATURE SELECTION CSV
# ======================================================================

feature_selection_df = (

    pd.DataFrame(
        {

            "Feature":
                feature_columns,

            "PSO_Selected":
                [
                    int(value)

                    for value
                    in best_particle
                ],

            "Status":
                [
                    "Selected"
                    if value == 1
                    else
                    "Rejected"

                    for value
                    in best_particle
                ]
        }
    )
)


feature_selection_df.to_csv(

    FEATURE_SELECTION_CSV_PATH,

    index=False
)


print(
    "\nSaved:",
    FEATURE_SELECTION_CSV_PATH
)


# ======================================================================
# 32. PSO FEATURE SELECTION GRAPH
# ======================================================================

plt.figure(

    figsize=(

        12,

        max(
            6,
            len(
                feature_columns
            )
            * 0.55
        )
    )
)


y_positions = (

    np.arange(
        len(
            feature_columns
        )
    )
)


plt.barh(

    y_positions,

    best_particle
)


plt.yticks(

    y_positions,

    feature_columns
)


plt.xticks(

    [0, 1],

    [
        "Rejected",
        "Selected"
    ]
)


plt.xlabel(
    "PSO Selection Status"
)


plt.ylabel(
    "Feature"
)


plt.title(

    "PSO Feature Selection",

    fontsize=14,

    fontweight="bold"
)


plt.grid(

    axis="x",

    alpha=0.25
)


plt.tight_layout()


plt.savefig(

    FEATURE_SELECTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    FEATURE_SELECTION_GRAPH_PATH
)


# ======================================================================
# 33. PSO FITNESS CONVERGENCE GRAPH
# ======================================================================

iterations_axis = (

    np.arange(
        1,
        ITERATIONS + 1
    )
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    iterations_axis,

    global_best_history,

    marker="o",

    linewidth=2,

    label="Global Best Fitness"
)


plt.plot(

    iterations_axis,

    average_fitness_history,

    marker="s",

    linewidth=2,

    label="Average Swarm Fitness"
)


plt.xlabel(
    "PSO Iteration"
)


plt.ylabel(
    "Fitness"
)


plt.title(

    "PSO Fitness Convergence",

    fontsize=14,

    fontweight="bold"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(

    FITNESS_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    FITNESS_GRAPH_PATH
)


# ======================================================================
# 34. CREATE FINAL PSO FEATURE DATAFRAME
# ======================================================================

X_selected_dataframe = (

    df[
        selected_features
    ]

    .copy()
)


X_selected_dataframe = (

    X_selected_dataframe
    .replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )
)


# ======================================================================
# 35. FINAL IMPUTATION
# ======================================================================

final_imputer = (

    SimpleImputer(
        strategy="median"
    )
)


X_selected_imputed = (

    final_imputer
    .fit_transform(
        X_selected_dataframe
    )
)


# ======================================================================
# 36. FINAL SCALING
# ======================================================================

final_scaler = (

    StandardScaler()
)


X_selected_scaled = (

    final_scaler
    .fit_transform(
        X_selected_imputed
    )
)


print(
    "\nFinal PSO-selected feature matrix:",
    X_selected_scaled.shape
)


# ======================================================================
# 37. TRAIN / TEST SPLIT
# ======================================================================

unique_classes, split_counts = (

    np.unique(
        y,
        return_counts=True
    )
)


if (
    split_counts.min()
    >=
    2
):

    stratify_option = y


else:

    stratify_option = None


sample_indices = (

    np.arange(
        len(df)
    )
)


(
    X_train,
    X_test,
    y_train,
    y_test,
    train_indices,
    test_indices

) = train_test_split(

    X_selected_scaled,

    y,

    sample_indices,

    test_size=0.25,

    random_state=RANDOM_STATE,

    stratify=stratify_option
)


print(
    "\nTraining samples:",
    len(
        X_train
    )
)


print(
    "Testing samples:",
    len(
        X_test
    )
)


# ======================================================================
# 38. BUILD PSO-ANN MODEL
# ======================================================================

number_of_features = (

    X_train.shape[1]
)


number_of_classes = (

    len(
        class_names
    )
)


ann_model = (

    Sequential(
        [

            tf.keras.Input(

                shape=(
                    number_of_features,
                )
            ),


            Dense(

                64,

                activation="relu"
            ),


            BatchNormalization(),


            Dropout(
                0.25
            ),


            Dense(

                32,

                activation="relu"
            ),


            Dropout(
                0.20
            ),


            Dense(

                16,

                activation="relu"
            ),


            Dense(

                number_of_classes,

                activation="softmax"
            )
        ]
    )
)


# ======================================================================
# 39. COMPILE ANN
# ======================================================================

ann_model.compile(

    optimizer=(

        tf.keras.optimizers.Adam(
            learning_rate=0.001
        )
    ),

    loss=(
        "sparse_categorical_crossentropy"
    ),

    metrics=[
        "accuracy"
    ]
)


print(
    "\nPSO-ANN Model:"
)


ann_model.summary()


# ======================================================================
# 40. EARLY STOPPING
# ======================================================================

early_stopping = (

    EarlyStopping(

        monitor="val_loss",

        patience=20,

        restore_best_weights=True,

        verbose=1
    )
)


# ======================================================================
# 41. TRAIN PSO-ANN
# ======================================================================

print(
    "\n"
    + "=" * 80
)

print(
    "TRAINING PSO-SELECTED ANN"
)

print(
    "=" * 80
)


history = (

    ann_model.fit(

        X_train,

        y_train,

        validation_split=0.20,

        epochs=200,

        batch_size=4,

        callbacks=[
            early_stopping
        ],

        verbose=1
    )
)


# ======================================================================
# 42. ANN PREDICTIONS
# ======================================================================

ann_probabilities = (

    ann_model.predict(

        X_test,

        verbose=0
    )
)


ann_predictions = (

    np.argmax(

        ann_probabilities,

        axis=1
    )
)


ann_accuracy = (

    accuracy_score(

        y_test,

        ann_predictions
    )
)


print(

    "\nPSO-ANN Accuracy:",

    f"{ann_accuracy * 100:.2f}%"
)


# ======================================================================
# 43. SAVE H5 MODEL
# ======================================================================

ann_model.save(
    H5_PATH
)


print(
    "\nSaved:",
    H5_PATH
)


# ======================================================================
# 44. SAVE PKL PREPROCESSOR
# ======================================================================

preprocessor = {

    "optimization_method":
        "Particle Swarm Optimization",

    "algorithm":
        "Binary Particle Swarm Optimization",

    "original_feature_columns":
        feature_columns,

    "selected_features":
        selected_features,

    "rejected_features":
        rejected_features,

    "best_particle":
        best_particle.tolist(),

    "best_fitness":
        float(
            global_best_fitness
        ),

    "imputer":
        final_imputer,

    "scaler":
        final_scaler,

    "label_encoder":
        label_encoder,

    "class_names":
        class_names,

    "target_source_column":
        target_source_column,

    "state_column":
        state_column,

    "lower_threshold":
        float(
            lower_threshold
        ),

    "upper_threshold":
        float(
            upper_threshold
        )
}


with open(

    PKL_PATH,

    "wb"

) as file:


    pickle.dump(

        preprocessor,

        file
    )


print(
    "Saved:",
    PKL_PATH
)


# ======================================================================
# 45. PSO ANN ACCURACY GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    history.history[
        "accuracy"
    ],

    linewidth=2,

    label="Training Accuracy"
)


plt.plot(

    history.history[
        "val_accuracy"
    ],

    linewidth=2,

    label="Validation Accuracy"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.title(

    "PSO-ANN Training and Validation Accuracy",

    fontsize=14,

    fontweight="bold"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    ACCURACY_GRAPH_PATH
)


# ======================================================================
# 46. PSO CORRELATION HEATMAP
# ======================================================================

heatmap_columns = (

    selected_features

    +

    [
        target_source_column
    ]
)


# Remove duplicate column names if present.

heatmap_columns = (

    list(
        dict.fromkeys(
            heatmap_columns
        )
    )
)


correlation_matrix = (

    df[
        heatmap_columns
    ]

    .corr()
)


fig, ax = plt.subplots(

    figsize=(

        max(
            9,
            len(
                heatmap_columns
            )
            * 1.2
        ),

        max(
            7,
            len(
                heatmap_columns
            )
            * 1.0
        )
    )
)


heatmap_image = (

    ax.imshow(

        correlation_matrix,

        aspect="auto"
    )
)


fig.colorbar(

    heatmap_image,

    ax=ax,

    label="Correlation"
)


ax.set_xticks(

    np.arange(
        len(
            correlation_matrix.columns
        )
    )
)


ax.set_yticks(

    np.arange(
        len(
            correlation_matrix.columns
        )
    )
)


ax.set_xticklabels(

    correlation_matrix.columns,

    rotation=45,

    ha="right",

    fontsize=8
)


ax.set_yticklabels(

    correlation_matrix.columns,

    fontsize=8
)


for row_index in range(

    len(
        correlation_matrix.columns
    )
):


    for column_index in range(

        len(
            correlation_matrix.columns
        )
    ):


        correlation_value = (

            correlation_matrix.iloc[
                row_index,
                column_index
            ]
        )


        if not np.isnan(
            correlation_value
        ):


            ax.text(

                column_index,

                row_index,

                f"{correlation_value:.2f}",

                ha="center",

                va="center",

                fontsize=7
            )


ax.set_title(

    "PSO-Selected Feature Correlation Heatmap",

    fontsize=14,

    fontweight="bold",

    pad=15
)


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    HEATMAP_PATH
)


# ======================================================================
# 47. DEFINE MACHINE LEARNING MODELS
# ======================================================================

print(
    "\n"
    + "=" * 80
)

print(
    "PSO-SELECTED MODEL COMPARISON"
)

print(
    "=" * 80
)


models = {


    "Logistic Regression":

        LogisticRegression(

            max_iter=2000,

            random_state=RANDOM_STATE
        ),


    "Decision Tree":

        DecisionTreeClassifier(

            random_state=RANDOM_STATE
        ),


    "Random Forest":

        RandomForestClassifier(

            n_estimators=200,

            random_state=RANDOM_STATE,

            class_weight="balanced"
        ),


    "KNN":

        KNeighborsClassifier(

            n_neighbors=3
        ),


    "SVM":

        SVC(

            kernel="rbf",

            probability=True,

            random_state=RANDOM_STATE
        ),


    "Gradient Boosting":

        GradientBoostingClassifier(

            random_state=RANDOM_STATE
        )
}


# ======================================================================
# 48. TRAIN COMPARISON MODELS
# ======================================================================

model_names = []

model_accuracies = []

model_predictions = {}


for (
    model_name,
    classifier

) in models.items():


    print(
        "\nTraining:",
        model_name
    )


    classifier.fit(

        X_train,

        y_train
    )


    predictions = (

        classifier.predict(
            X_test
        )
    )


    accuracy = (

        accuracy_score(

            y_test,

            predictions
        )
    )


    model_names.append(
        model_name
    )


    model_accuracies.append(
        accuracy
    )


    model_predictions[
        model_name
    ] = predictions


    print(

        f"{model_name} Accuracy: "

        f"{accuracy * 100:.2f}%"
    )


# Add PSO ANN result

model_names.append(
    "Artificial Neural Network"
)


model_accuracies.append(
    ann_accuracy
)


model_predictions[
    "Artificial Neural Network"
] = ann_predictions


# ======================================================================
# 49. PSO MODEL COMPARISON GRAPH
# ======================================================================

accuracy_percentages = [

    value * 100

    for value
    in model_accuracies
]


plt.figure(
    figsize=(12, 7)
)


bars = plt.bar(

    model_names,

    accuracy_percentages
)


plt.xlabel(
    "Classification Model"
)


plt.ylabel(
    "Accuracy (%)"
)


plt.title(

    "PSO-Selected Feature Model Accuracy Comparison",

    fontsize=14,

    fontweight="bold"
)


plt.xticks(

    rotation=35,

    ha="right"
)


plt.ylim(
    0,
    110
)


for (
    bar,
    accuracy_percentage

) in zip(

    bars,

    accuracy_percentages
):


    plt.text(

        bar.get_x()
        +
        bar.get_width()
        / 2,

        accuracy_percentage
        + 1,

        f"{accuracy_percentage:.2f}%",

        ha="center",

        va="bottom",

        fontsize=9
    )


plt.grid(

    axis="y",

    alpha=0.25
)


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


# ======================================================================
# 50. CREATE PSO RESULT CSV
# ======================================================================

result_df = (

    pd.DataFrame(
        {

            "Model":
                model_names,

            "Accuracy":
                model_accuracies,

            "Accuracy_Percentage":
                [

                    round(
                        value * 100,
                        2
                    )

                    for value
                    in model_accuracies
                ],

            "PSO_Features_Selected":
                len(
                    selected_features
                ),

            "Total_Candidate_Features":
                NUMBER_OF_FEATURES,

            "PSO_Best_Fitness":
                round(
                    global_best_fitness,
                    6
                )
        }
    )
)


result_df = (

    result_df

    .sort_values(

        by="Accuracy",

        ascending=False
    )

    .reset_index(
        drop=True
    )
)


result_df.insert(

    0,

    "Rank",

    np.arange(

        1,

        len(
            result_df
        )
        +
        1
    )
)


result_df.to_csv(

    RESULT_CSV_PATH,

    index=False
)


print(
    "\nPSO RESULT CSV:"
)


print(
    result_df
)


print(
    "\nSaved:",
    RESULT_CSV_PATH
)


# ======================================================================
# 51. PSO RESULT GRAPH
# ======================================================================

plt.figure(
    figsize=(12, 7)
)


bars = plt.barh(

    result_df[
        "Model"
    ],

    result_df[
        "Accuracy_Percentage"
    ]
)


plt.xlabel(
    "Accuracy (%)"
)


plt.ylabel(
    "Classification Model"
)


plt.title(

    "PSO-Based Classification Results",

    fontsize=14,

    fontweight="bold"
)


plt.xlim(
    0,
    110
)


for (
    bar,
    value

) in zip(

    bars,

    result_df[
        "Accuracy_Percentage"
    ]
):


    plt.text(

        value + 1,

        bar.get_y()
        +
        bar.get_height()
        / 2,

        f"{value:.2f}%",

        va="center"
    )


plt.grid(

    axis="x",

    alpha=0.25
)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


# ======================================================================
# 52. CREATE PSO PREDICTION CSV
# ======================================================================

actual_labels = (

    label_encoder
    .inverse_transform(
        y_test
    )
)


predicted_labels = (

    label_encoder
    .inverse_transform(
        ann_predictions
    )
)


prediction_df = (
    pd.DataFrame()
)


# Add State / Region

if (

    state_column is not None

    and

    state_column
    in
    df.columns
):


    prediction_df[
        "State_Region"
    ] = (

        df.iloc[
            test_indices
        ][
            state_column
        ]

        .astype(str)

        .values
    )


else:


    prediction_df[
        "Sample"
    ] = [

        f"Sample_{index + 1}"

        for index in range(
            len(
                y_test
            )
        )
    ]


# Add actual numeric unmet need

prediction_df[
    "Actual_Unmet_Need"
] = (

    df.iloc[
        test_indices
    ][
        target_source_column
    ]

    .values
)


# Actual category

prediction_df[
    "Actual_Category"
] = (
    actual_labels
)


# Predicted category

prediction_df[
    "PSO_Predicted_Category"
] = (
    predicted_labels
)


# Correct / incorrect prediction

prediction_df[
    "Correct_Prediction"
] = (

    prediction_df[
        "Actual_Category"
    ]

    ==

    prediction_df[
        "PSO_Predicted_Category"
    ]
)


# Prediction confidence

prediction_df[
    "PSO_Prediction_Confidence"
] = (

    np.max(

        ann_probabilities,

        axis=1
    )
)


# ======================================================================
# 53. ADD CLASS PROBABILITIES
# ======================================================================

for (
    class_index,
    class_name

) in enumerate(
    class_names
):


    prediction_df[
        f"PSO_Probability_{class_name}"
    ] = (

        ann_probabilities[
            :,
            class_index
        ]
    )


# ======================================================================
# 54. SAVE PREDICTION CSV
# ======================================================================

prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False
)


print(
    "\nPSO PREDICTIONS:"
)


print(
    prediction_df
)


print(
    "\nSaved:",
    PREDICTION_CSV_PATH
)


# ======================================================================
# 55. PSO PREDICTION GRAPH
# ======================================================================

if (
    "State_Region"
    in
    prediction_df.columns
):


    sample_labels = (

        prediction_df[
            "State_Region"
        ]

        .astype(str)

        .tolist()
    )


else:


    sample_labels = (

        prediction_df[
            "Sample"
        ]

        .astype(str)

        .tolist()
    )


x_positions = (

    np.arange(
        len(
            prediction_df
        )
    )
)


plt.figure(

    figsize=(

        max(

            12,

            len(
                prediction_df
            )
            *
            1.2
        ),

        7
    )
)


plt.plot(

    x_positions,

    y_test,

    marker="o",

    linewidth=2,

    markersize=8,

    label="Actual Category"
)


plt.plot(

    x_positions,

    ann_predictions,

    marker="s",

    linestyle="--",

    linewidth=2,

    markersize=7,

    label="PSO-ANN Prediction"
)


plt.xticks(

    x_positions,

    sample_labels,

    rotation=45,

    ha="right"
)


plt.yticks(

    np.arange(
        len(
            class_names
        )
    ),

    class_names
)


plt.xlabel(
    "State / Test Sample"
)


plt.ylabel(
    "Unmet Need Category"
)


plt.title(

    "PSO-ANN Actual vs Predicted Unmet Need Category",

    fontsize=14,

    fontweight="bold"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


# ======================================================================
# 56. CONFUSION MATRIX
# ======================================================================

confusion = (

    confusion_matrix(

        y_test,

        ann_predictions,

        labels=np.arange(
            len(
                class_names
            )
        )
    )
)


print(
    "\n"
    + "=" * 80
)


print(
    "PSO-ANN CONFUSION MATRIX"
)


print(
    "=" * 80
)


print(
    confusion
)


# ======================================================================
# 57. CLASSIFICATION REPORT
# ======================================================================

classification_report_text = (

    classification_report(

        y_test,

        ann_predictions,

        labels=np.arange(
            len(
                class_names
            )
        ),

        target_names=class_names,

        zero_division=0
    )
)


print(
    "\nClassification Report:\n"
)


print(
    classification_report_text
)


classification_report_dictionary = (

    classification_report(

        y_test,

        ann_predictions,

        labels=np.arange(
            len(
                class_names
            )
        ),

        target_names=class_names,

        zero_division=0,

        output_dict=True
    )
)


# ======================================================================
# 58. FIND BEST MODEL
# ======================================================================

best_model_index = (

    int(
        np.argmax(
            model_accuracies
        )
    )
)


best_model_name = (

    model_names[
        best_model_index
    ]
)


best_model_accuracy = (

    model_accuracies[
        best_model_index
    ]
)


print(
    "\n"
    + "=" * 80
)


print(
    "BEST PSO-BASED MODEL"
)


print(
    "=" * 80
)


print(
    "\nModel:",
    best_model_name
)


print(

    "Accuracy:",

    f"{best_model_accuracy * 100:.2f}%"
)


# ======================================================================
# 59. SAVE YAML CONFIGURATION
# ======================================================================

config = {


    "project": {

        "name":
            "State-Level Family Planning Unmet Need Pattern Analysis and Classification",

        "optimization_method":
            "Particle Swarm Optimization",

        "dataset":
            DATASET_PATH,

        "output_directory":
            PROJECT_DIR
    },


    "pso": {

        "algorithm":
            "Binary Particle Swarm Optimization",

        "swarm_size":
            SWARM_SIZE,

        "iterations":
            ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,

        "social_coefficient":
            SOCIAL_COEFFICIENT,

        "maximum_velocity":
            MAX_VELOCITY,

        "feature_penalty":
            FEATURE_PENALTY,

        "best_fitness":
            float(
                global_best_fitness
            ),

        "candidate_feature_count":
            int(
                NUMBER_OF_FEATURES
            ),

        "selected_feature_count":
            int(
                len(
                    selected_features
                )
            ),

        "selected_features":
            [
                str(feature)

                for feature
                in selected_features
            ],

        "rejected_features":
            [
                str(feature)

                for feature
                in rejected_features
            ],

        "best_particle":
            [
                int(value)

                for value
                in best_particle
            ]
    },


    "classification": {

        "target_source":
            str(
                target_source_column
            ),

        "target":
            "Unmet_Need_Category",

        "classes":
            [
                str(class_name)

                for class_name
                in class_names
            ],

        "lower_threshold":
            float(
                lower_threshold
            ),

        "upper_threshold":
            float(
                upper_threshold
            )
    },


    "ann": {

        "input_features":
            int(
                len(
                    selected_features
                )
            ),

        "hidden_layers":
            [
                64,
                32,
                16
            ],

        "optimizer":
            "Adam",

        "learning_rate":
            0.001,

        "loss":
            "Sparse Categorical Crossentropy",

        "maximum_epochs":
            200,

        "batch_size":
            4,

        "test_accuracy":
            float(
                ann_accuracy
            )
    },


    "best_model": {

        "name":
            str(
                best_model_name
            ),

        "accuracy":
            float(
                best_model_accuracy
            )
    }
}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:


    yaml.safe_dump(

        config,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nSaved:",
    YAML_PATH
)


# ======================================================================
# 60. SAVE JSON METADATA
# ======================================================================

metadata = {


    "project_name":
        "State-Level Family Planning Unmet Need Pattern Analysis and Classification",


    "optimization_method":
        "Particle Swarm Optimization",


    "pso_algorithm":
        "Binary Particle Swarm Optimization",


    "dataset_path":
        DATASET_PATH,


    "dataset_rows":
        int(
            len(df)
        ),


    "dataset_columns":
        int(
            len(
                df.columns
            )
        ),


    "original_columns":
        [
            str(column)

            for column
            in original_columns
        ],


    "state_column":
        str(
            state_column
        ),


    "target_source_column":
        str(
            target_source_column
        ),


    "target_column":
        "Unmet_Need_Category",


    "class_names":
        [
            str(class_name)

            for class_name
            in class_names
        ],


    "candidate_features":
        [
            str(feature)

            for feature
            in feature_columns
        ],


    "selected_features":
        [
            str(feature)

            for feature
            in selected_features
        ],


    "rejected_features":
        [
            str(feature)

            for feature
            in rejected_features
        ],


    "best_particle":
        [
            int(value)

            for value
            in best_particle
        ],


    "best_pso_fitness":
        float(
            global_best_fitness
        ),


    "pso_parameters": {

        "swarm_size":
            SWARM_SIZE,

        "iterations":
            ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,

        "social_coefficient":
            SOCIAL_COEFFICIENT,

        "feature_penalty":
            FEATURE_PENALTY
    },


    "training_samples":
        int(
            len(
                X_train
            )
        ),


    "testing_samples":
        int(
            len(
                X_test
            )
        ),


    "ann_accuracy":
        float(
            ann_accuracy
        ),


    "best_model":
        str(
            best_model_name
        ),


    "best_model_accuracy":
        float(
            best_model_accuracy
        ),


    "confusion_matrix":
        confusion.tolist(),


    "classification_report":
        classification_report_dictionary
}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:


    json.dump(

        metadata,

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "Saved:",
    JSON_PATH
)


# ======================================================================
# 61. VERIFY H5 MODEL
# ======================================================================

print(
    "\n"
    + "=" * 80
)


print(
    "VERIFYING PSO H5 MODEL"
)


print(
    "=" * 80
)


loaded_model = (

    tf.keras.models.load_model(
        H5_PATH
    )
)


verification_probabilities = (

    loaded_model.predict(

        X_test[
            :1
        ],

        verbose=0
    )
)


verification_prediction = (

    int(

        np.argmax(

            verification_probabilities,

            axis=1
        )[0]
    )
)


verification_label = (

    label_encoder
    .inverse_transform(
        [
            verification_prediction
        ]
    )[0]
)


print(
    "\nH5 model loaded successfully."
)


print(
    "Sample predicted category:",
    verification_label
)


# ======================================================================
# 62. VERIFY PKL PREPROCESSOR
# ======================================================================

print(
    "\n"
    + "=" * 80
)


print(
    "VERIFYING PSO PKL"
)


print(
    "=" * 80
)


with open(

    PKL_PATH,

    "rb"

) as file:


    loaded_preprocessor = (

        pickle.load(
            file
        )
    )


print(
    "\nPKL loaded successfully."
)


print(
    "\nSaved PSO-selected features:"
)


for feature in (

    loaded_preprocessor[
        "selected_features"
    ]
):

    print(
        " -",
        feature
    )


# ======================================================================
# 63. VERIFY GENERATED FILES
# ======================================================================

output_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_CSV_PATH,

    FEATURE_SELECTION_GRAPH_PATH,

    FITNESS_GRAPH_PATH
]


print(
    "\n"
    + "=" * 80
)


print(
    "VERIFYING ALL PSO OUTPUT FILES"
)


print(
    "=" * 80
)


for file_path in output_files:


    if os.path.exists(
        file_path
    ):


        file_size_kb = (

            os.path.getsize(
                file_path
            )

            /

            1024
        )


        print(

            f"\n[OK] "

            f"{os.path.basename(file_path)}"
        )


        print(

            f"     Size: "

            f"{file_size_kb:.2f} KB"
        )


    else:


        print(

            f"\n[ERROR] "

            f"{os.path.basename(file_path)}"
        )


# ======================================================================
# 64. FINAL PSO SUMMARY
# ======================================================================

print(
    "\n"
    + "=" * 80
)


print(
    "PSO PROJECT SUMMARY"
)


print(
    "=" * 80
)


print(
    f"""

PROJECT
-------
State-Level Family Planning Unmet Need Pattern Analysis and Classification


OPTIMIZATION METHOD
-------------------
Particle Swarm Optimization


PSO ALGORITHM
-------------
Binary Particle Swarm Optimization


DATASET
-------
{DATASET_PATH}


TARGET SOURCE
-------------
{target_source_column}


CLASSIFICATION
--------------
Low
Moderate
High


PSO CONFIGURATION
-----------------
Swarm Size:
{SWARM_SIZE}

Iterations:
{ITERATIONS}

Inertia Weight:
{INERTIA_WEIGHT}

Cognitive Coefficient:
{COGNITIVE_COEFFICIENT}

Social Coefficient:
{SOCIAL_COEFFICIENT}

Feature Penalty:
{FEATURE_PENALTY}


PSO FEATURE SELECTION
---------------------
Candidate Features:
{NUMBER_OF_FEATURES}

Selected Features:
{len(selected_features)}

Rejected Features:
{len(rejected_features)}

Best PSO Fitness:
{global_best_fitness:.4f}


SELECTED FEATURES
-----------------
"""
)


for feature in selected_features:

    print(
        " -",
        feature
    )


print(
    f"""

MODEL PERFORMANCE
-----------------
PSO-ANN Accuracy:
{ann_accuracy * 100:.2f}%

Best Classification Model:
{best_model_name}

Best Accuracy:
{best_model_accuracy * 100:.2f}%


GENERATED FILES
===============


MODEL / CONFIGURATION FILES
---------------------------

1. pso_family_planning_model.h5

2. pso_family_planning_preprocessor.pkl

3. pso_family_planning_config.yaml

4. pso_family_planning_metadata.json


PSO FEATURE SELECTION
---------------------

5. pso_feature_selection.csv

6. pso_feature_selection_graph.png

7. pso_fitness_graph.png


MODEL VISUALIZATIONS
--------------------

8. pso_accuracy_graph.png

9. pso_heatmap.png

10. pso_comparison_graph.png


RESULT FILES
------------

11. pso_result.csv

12. pso_result_graph.png


PREDICTION FILES
----------------

13. pso_prediction.csv

14. pso_prediction_graph.png


OUTPUT DIRECTORY
----------------
{PROJECT_DIR}

"""
)


print(
    "=" * 80
)


print(
    "PSO EXECUTION COMPLETED SUCCESSFULLY"
)


print(
    "=" * 80
)
```
