```python
import os
import json
import pickle
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yaml

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Dense,
    Dropout,
    BatchNormalization
)
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC


warnings.filterwarnings("ignore")


# ======================================================================
# 2. RANDOM SEEDS
# ======================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# ======================================================================
# 3. PROJECT PATH
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\State-Level Family Planning Unmet Need Pattern Analysis and Classification"
)

DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "unmetneedforFPNFHS3.csv"
)

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# 4. OUTPUT PATHS
# ======================================================================

H5_PATH = os.path.join(
    PROJECT_DIR,
    "ais_family_planning_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "ais_family_planning_preprocessor.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "ais_family_planning_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "ais_family_planning_metadata.json"
)


ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_accuracy_graph.png"
)

HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "ais_heatmap.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_comparison_graph.png"
)

RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction_graph.png"
)

FEATURE_SELECTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_feature_selection.csv"
)

FEATURE_SELECTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_feature_selection_graph.png"
)

FITNESS_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_fitness_graph.png"
)


# ======================================================================
# 5. LOAD DATASET
# ======================================================================

print("=" * 80)
print("AIS-BASED FAMILY PLANNING UNMET NEED ANALYSIS")
print("=" * 80)

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}"
    )


df = pd.read_csv(
    DATASET_PATH
)


print("\nDataset loaded successfully.")

print(
    "Original shape:",
    df.shape
)


print("\nOriginal columns:")

for i, col in enumerate(
    df.columns,
    1
):

    print(
        f"{i}. {col}"
    )


# ======================================================================
# 6. CLEAN COLUMN NAMES
# ======================================================================

original_columns = (
    df.columns
    .astype(str)
    .tolist()
)


df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        "\n",
        " ",
        regex=False
    )
    .str.replace(
        "\r",
        " ",
        regex=False
    )
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


df = df.dropna(
    axis=1,
    how="all"
)


print(
    "\nShape after cleaning:",
    df.shape
)


# ======================================================================
# 7. DETECT STATE / REGION COLUMN
# ======================================================================

state_column = None


for col in df.columns:

    lower = col.lower()

    if (
        "state" in lower
        or "region" in lower
        or "area" in lower
    ):

        state_column = col
        break


if state_column is None:

    object_columns = (
        df.select_dtypes(
            include=[
                "object",
                "string"
            ]
        )
        .columns
        .tolist()
    )

    if len(object_columns) > 0:

        state_column = (
            object_columns[0]
        )


print(
    "\nDetected State/Region column:",
    state_column
)


# ======================================================================
# 8. CONVERT POSSIBLE NUMERIC COLUMNS
# ======================================================================

for col in df.columns:

    if col == state_column:
        continue


    if df[col].dtype == "object":

        cleaned = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(
                ",",
                "",
                regex=False
            )
            .str.replace(
                "%",
                "",
                regex=False
            )
        )


        numeric_version = (
            pd.to_numeric(
                cleaned,
                errors="coerce"
            )
        )


        original_non_null = (
            df[col]
            .notna()
            .sum()
        )

        numeric_non_null = (
            numeric_version
            .notna()
            .sum()
        )


        if (
            original_non_null > 0
            and
            numeric_non_null
            / original_non_null
            >= 0.60
        ):

            df[col] = (
                numeric_version
            )


# ======================================================================
# 9. NUMERIC COLUMNS
# ======================================================================

numeric_columns = (
    df.select_dtypes(
        include=[
            np.number
        ]
    )
    .columns
    .tolist()
)


print("\nNumeric columns:")

for col in numeric_columns:

    print(
        " -",
        col
    )


if len(numeric_columns) < 2:

    raise ValueError(
        "At least two numeric columns are required."
    )


# ======================================================================
# 10. FIND TARGET SOURCE COLUMN
# ======================================================================

target_source_column = None


# First preference:
# unmet + total

for col in numeric_columns:

    lower = col.lower()

    if (
        "unmet" in lower
        and
        "total" in lower
    ):

        target_source_column = col
        break


# Second preference:
# total

if target_source_column is None:

    for col in numeric_columns:

        if "total" in col.lower():

            target_source_column = col
            break


# Fallback

if target_source_column is None:

    target_source_column = (
        numeric_columns[0]
    )


print(
    "\nTarget source column:"
)

print(
    target_source_column
)


# ======================================================================
# 11. REMOVE MISSING TARGET ROWS
# ======================================================================

df = (
    df.dropna(
        subset=[
            target_source_column
        ]
    )
    .reset_index(
        drop=True
    )
)


# ======================================================================
# 12. CREATE TARGET CATEGORIES
# ======================================================================

target_values = (
    df[
        target_source_column
    ]
    .astype(float)
)


lower_threshold = (
    target_values
    .quantile(
        1 / 3
    )
)

upper_threshold = (
    target_values
    .quantile(
        2 / 3
    )
)


def create_category(value):

    if value <= lower_threshold:

        return "Low"

    elif value <= upper_threshold:

        return "Moderate"

    else:

        return "High"


df[
    "Unmet_Need_Category"
] = (
    target_values
    .apply(
        create_category
    )
)


print("\nTarget thresholds:")

print(
    f"Low <= "
    f"{lower_threshold:.3f}"
)

print(
    f"Moderate > "
    f"{lower_threshold:.3f} "
    f"and <= "
    f"{upper_threshold:.3f}"
)

print(
    f"High > "
    f"{upper_threshold:.3f}"
)


print("\nTarget distribution:")

print(
    df[
        "Unmet_Need_Category"
    ]
    .value_counts()
)


# ======================================================================
# 13. LABEL ENCODER
# ======================================================================

label_encoder = (
    LabelEncoder()
)


y = (
    label_encoder
    .fit_transform(
        df[
            "Unmet_Need_Category"
        ]
        .astype(str)
    )
)


class_names = (
    label_encoder
    .classes_
    .tolist()
)


print("\nClass mapping:")

for index, name in enumerate(
    class_names
):

    print(
        index,
        "=",
        name
    )


# ======================================================================
# 14. PREPARE ORIGINAL FEATURE SET
# ======================================================================

# The target source MUST NOT be used as a predictor because the
# classification target is generated directly from it.

feature_columns = [

    col

    for col in numeric_columns

    if col != target_source_column
]


if len(feature_columns) == 0:

    raise ValueError(
        "No predictor features available."
    )


print("\nCandidate features for AIS:")

for feature in feature_columns:

    print(
        " -",
        feature
    )


X_dataframe = (
    df[
        feature_columns
    ]
    .copy()
)


X_dataframe = (
    X_dataframe
    .replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )
)


# ======================================================================
# 15. IMPUTATION
# ======================================================================

imputer = (
    SimpleImputer(
        strategy="median"
    )
)


X_imputed = (
    imputer
    .fit_transform(
        X_dataframe
    )
)


# ======================================================================
# 16. SCALE ORIGINAL FEATURE MATRIX
# ======================================================================

initial_scaler = (
    StandardScaler()
)


X_all_scaled = (
    initial_scaler
    .fit_transform(
        X_imputed
    )
)


print(
    "\nInitial feature matrix:",
    X_all_scaled.shape
)


# ======================================================================
# 17. ARTIFICIAL IMMUNE SYSTEM CONFIGURATION
# ======================================================================

print("\n" + "=" * 80)
print("ARTIFICIAL IMMUNE SYSTEM FEATURE SELECTION")
print("=" * 80)


NUMBER_OF_FEATURES = (
    len(feature_columns)
)

POPULATION_SIZE = 30

GENERATIONS = 30

ELITE_SIZE = 8

CLONES_PER_ELITE = 5

MUTATION_RATE = 0.20

RANDOM_NEW_ANTIBODIES = 5

FEATURE_PENALTY = 0.02


print(
    "\nAIS Configuration:"
)

print(
    "Population Size:",
    POPULATION_SIZE
)

print(
    "Generations:",
    GENERATIONS
)

print(
    "Elite Antibodies:",
    ELITE_SIZE
)

print(
    "Clones Per Elite:",
    CLONES_PER_ELITE
)

print(
    "Mutation Rate:",
    MUTATION_RATE
)

print(
    "Number of Candidate Features:",
    NUMBER_OF_FEATURES
)


# ======================================================================
# 18. AIS HELPER FUNCTION:
# CREATE RANDOM ANTIBODY
# ======================================================================

def create_random_antibody(
    number_of_features
):

    antibody = (
        np.random.randint(
            0,
            2,
            size=number_of_features
        )
    )


    # At least one feature must be selected.

    if antibody.sum() == 0:

        random_index = (
            np.random.randint(
                0,
                number_of_features
            )
        )

        antibody[
            random_index
        ] = 1


    return antibody


# ======================================================================
# 19. CROSS-VALIDATION SETUP FOR AIS
# ======================================================================

class_counts = (
    np.bincount(y)
)

minimum_class_count = (
    class_counts.min()
)


if minimum_class_count >= 3:

    CV_SPLITS = 3

elif minimum_class_count >= 2:

    CV_SPLITS = 2

else:

    CV_SPLITS = None


if CV_SPLITS is None:

    print(
        "\nWarning: Insufficient samples for "
        "stratified cross-validation."
    )

else:

    print(
        "\nAIS cross-validation folds:",
        CV_SPLITS
    )


# ======================================================================
# 20. AIS FITNESS FUNCTION
# ======================================================================

fitness_cache = {}


def calculate_fitness(
    antibody
):

    # Convert mask into tuple for cache.

    key = tuple(
        antibody.tolist()
    )


    if key in fitness_cache:

        return fitness_cache[key]


    selected_indices = (
        np.where(
            antibody == 1
        )[0]
    )


    if len(selected_indices) == 0:

        return 0.0


    X_selected = (
        X_all_scaled[
            :,
            selected_indices
        ]
    )


    classifier = (
        RandomForestClassifier(
            n_estimators=100,
            random_state=RANDOM_STATE,
            class_weight="balanced"
        )
    )


    try:

        if CV_SPLITS is not None:

            cv = (
                StratifiedKFold(
                    n_splits=CV_SPLITS,
                    shuffle=True,
                    random_state=RANDOM_STATE
                )
            )


            scores = (
                cross_val_score(
                    classifier,
                    X_selected,
                    y,
                    cv=cv,
                    scoring="accuracy"
                )
            )


            mean_accuracy = (
                scores.mean()
            )

        else:

            X_train_temp, X_test_temp, y_train_temp, y_test_temp = (
                train_test_split(
                    X_selected,
                    y,
                    test_size=0.25,
                    random_state=RANDOM_STATE
                )
            )


            classifier.fit(
                X_train_temp,
                y_train_temp
            )


            mean_accuracy = (
                classifier.score(
                    X_test_temp,
                    y_test_temp
                )
            )


        feature_ratio = (
            len(selected_indices)
            / NUMBER_OF_FEATURES
        )


        # Encourage high accuracy while mildly rewarding
        # smaller feature subsets.

        fitness = (
            mean_accuracy
            -
            FEATURE_PENALTY
            * feature_ratio
        )


    except Exception:

        fitness = 0.0


    fitness_cache[key] = (
        float(fitness)
    )


    return float(fitness)


# ======================================================================
# 21. MUTATION FUNCTION
# ======================================================================

def mutate_antibody(
    antibody,
    mutation_rate
):

    mutated = (
        antibody.copy()
    )


    for i in range(
        len(mutated)
    ):

        if (
            np.random.random()
            <
            mutation_rate
        ):

            mutated[i] = (
                1 - mutated[i]
            )


    # Prevent empty antibody.

    if mutated.sum() == 0:

        random_index = (
            np.random.randint(
                0,
                len(mutated)
            )
        )

        mutated[
            random_index
        ] = 1


    return mutated


# ======================================================================
# 22. INITIAL AIS POPULATION
# ======================================================================

population = [

    create_random_antibody(
        NUMBER_OF_FEATURES
    )

    for _ in range(
        POPULATION_SIZE
    )
]


best_fitness_history = []

average_fitness_history = []

best_antibody = None

best_fitness = -np.inf


# ======================================================================
# 23. RUN CLONAL SELECTION AIS
# ======================================================================

for generation in range(
    GENERATIONS
):

    evaluated_population = []


    for antibody in population:

        fitness = (
            calculate_fitness(
                antibody
            )
        )

        evaluated_population.append(
            (
                antibody,
                fitness
            )
        )


    # Sort descending by fitness.

    evaluated_population.sort(
        key=lambda x: x[1],
        reverse=True
    )


    generation_best_antibody = (
        evaluated_population[0][0]
        .copy()
    )

    generation_best_fitness = (
        evaluated_population[0][1]
    )


    generation_average_fitness = (
        np.mean(
            [
                item[1]
                for item in evaluated_population
            ]
        )
    )


    best_fitness_history.append(
        generation_best_fitness
    )

    average_fitness_history.append(
        generation_average_fitness
    )


    if (
        generation_best_fitness
        >
        best_fitness
    ):

        best_fitness = (
            generation_best_fitness
        )

        best_antibody = (
            generation_best_antibody
            .copy()
        )


    print(
        f"Generation "
        f"{generation + 1:02d}/{GENERATIONS} | "
        f"Best Fitness: "
        f"{generation_best_fitness:.4f} | "
        f"Average Fitness: "
        f"{generation_average_fitness:.4f} | "
        f"Selected Features: "
        f"{generation_best_antibody.sum()}"
    )


    # ================================================================
    # SELECT ELITE ANTIBODIES
    # ================================================================

    elite_population = (

        evaluated_population[
            :ELITE_SIZE
        ]
    )


    # ================================================================
    # CLONAL EXPANSION
    # ================================================================

    clones = []


    for rank, (
        elite_antibody,
        elite_fitness
    ) in enumerate(
        elite_population
    ):

        # Better antibodies mutate less.

        adaptive_mutation_rate = (

            MUTATION_RATE
            *
            (
                (rank + 1)
                /
                ELITE_SIZE
            )
        )


        for _ in range(
            CLONES_PER_ELITE
        ):

            clone = (
                mutate_antibody(
                    elite_antibody,
                    adaptive_mutation_rate
                )
            )

            clones.append(
                clone
            )


    # ================================================================
    # EVALUATE CLONES
    # ================================================================

    evaluated_clones = []


    for clone in clones:

        clone_fitness = (
            calculate_fitness(
                clone
            )
        )

        evaluated_clones.append(
            (
                clone,
                clone_fitness
            )
        )


    # ================================================================
    # COMBINE PARENTS + CLONES
    # ================================================================

    combined_population = (

        evaluated_population
        +
        evaluated_clones
    )


    combined_population.sort(
        key=lambda x: x[1],
        reverse=True
    )


    # Keep strongest antibodies.

    survivors_needed = (

        POPULATION_SIZE
        -
        RANDOM_NEW_ANTIBODIES
    )


    next_population = [

        item[0].copy()

        for item in combined_population[
            :survivors_needed
        ]
    ]


    # ================================================================
    # IMMUNE DIVERSITY
    # ================================================================

    for _ in range(
        RANDOM_NEW_ANTIBODIES
    ):

        next_population.append(

            create_random_antibody(
                NUMBER_OF_FEATURES
            )
        )


    population = (
        next_population
    )


# ======================================================================
# 24. AIS BEST FEATURE SUBSET
# ======================================================================

selected_indices = (
    np.where(
        best_antibody == 1
    )[0]
)


selected_features = [

    feature_columns[i]

    for i in selected_indices
]


rejected_features = [

    feature_columns[i]

    for i in range(
        NUMBER_OF_FEATURES
    )

    if best_antibody[i] == 0
]


print("\n" + "=" * 80)
print("AIS FEATURE SELECTION RESULT")
print("=" * 80)


print(
    "\nBest AIS Fitness:",
    round(
        best_fitness,
        4
    )
)


print(
    "\nTotal Candidate Features:",
    NUMBER_OF_FEATURES
)


print(
    "Selected Features:",
    len(
        selected_features
    )
)


print(
    "Rejected Features:",
    len(
        rejected_features
    )
)


print("\nAIS Selected Features:")

for feature in selected_features:

    print(
        " [SELECTED]",
        feature
    )


print("\nAIS Rejected Features:")

if len(rejected_features) == 0:

    print(
        " None"
    )

else:

    for feature in rejected_features:

        print(
            " [REJECTED]",
            feature
        )


# ======================================================================
# 25. SAVE AIS FEATURE SELECTION CSV
# ======================================================================

feature_selection_df = (
    pd.DataFrame(
        {
            "Feature":
                feature_columns,

            "AIS_Selected":
                [
                    int(value)
                    for value
                    in best_antibody
                ],

            "Status":
                [
                    "Selected"
                    if value == 1
                    else "Rejected"

                    for value
                    in best_antibody
                ]
        }
    )
)


feature_selection_df.to_csv(
    FEATURE_SELECTION_CSV_PATH,
    index=False
)


print(
    "\nSaved:",
    FEATURE_SELECTION_CSV_PATH
)


# ======================================================================
# 26. AIS FEATURE SELECTION GRAPH
# ======================================================================

plt.figure(
    figsize=(
        12,
        max(
            6,
            len(feature_columns)
            * 0.55
        )
    )
)


y_positions = (
    np.arange(
        len(feature_columns)
    )
)


plt.barh(
    y_positions,
    best_antibody
)


plt.yticks(
    y_positions,
    feature_columns
)


plt.xticks(
    [0, 1],
    [
        "Rejected",
        "Selected"
    ]
)


plt.xlabel(
    "AIS Selection Status"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "AIS Feature Selection",
    fontsize=14,
    fontweight="bold"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    FEATURE_SELECTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 27. AIS FITNESS GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


generations_axis = (
    np.arange(
        1,
        GENERATIONS + 1
    )
)


plt.plot(
    generations_axis,
    best_fitness_history,
    marker="o",
    linewidth=2,
    label="Best Fitness"
)


plt.plot(
    generations_axis,
    average_fitness_history,
    marker="s",
    linewidth=2,
    label="Average Fitness"
)


plt.xlabel(
    "AIS Generation"
)

plt.ylabel(
    "Fitness"
)

plt.title(
    "AIS Fitness Convergence",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


plt.savefig(
    FITNESS_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 28. CREATE AIS-SELECTED FEATURE MATRIX
# ======================================================================

X_selected_dataframe = (
    df[
        selected_features
    ]
    .copy()
)


X_selected_dataframe = (
    X_selected_dataframe
    .replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )
)


# ======================================================================
# 29. FINAL IMPUTER
# ======================================================================

final_imputer = (
    SimpleImputer(
        strategy="median"
    )
)


X_selected_imputed = (
    final_imputer
    .fit_transform(
        X_selected_dataframe
    )
)


# ======================================================================
# 30. FINAL SCALER
# ======================================================================

final_scaler = (
    StandardScaler()
)


X_selected_scaled = (
    final_scaler
    .fit_transform(
        X_selected_imputed
    )
)


print(
    "\nFinal AIS-selected feature matrix:",
    X_selected_scaled.shape
)


# ======================================================================
# 31. TRAIN / TEST SPLIT
# ======================================================================

unique_classes, split_counts = (
    np.unique(
        y,
        return_counts=True
    )
)


if split_counts.min() >= 2:

    stratify_option = y

else:

    stratify_option = None


indices = (
    np.arange(
        len(df)
    )
)


(
    X_train,
    X_test,
    y_train,
    y_test,
    train_indices,
    test_indices
) = train_test_split(

    X_selected_scaled,
    y,
    indices,

    test_size=0.25,

    random_state=RANDOM_STATE,

    stratify=stratify_option
)


print(
    "\nTraining Samples:",
    len(X_train)
)

print(
    "Testing Samples:",
    len(X_test)
)


# ======================================================================
# 32. BUILD AIS-BASED ANN
# ======================================================================

number_of_features = (
    X_train.shape[1]
)

number_of_classes = (
    len(class_names)
)


ann_model = Sequential([

    tf.keras.Input(
        shape=(
            number_of_features,
        )
    ),

    Dense(
        64,
        activation="relu"
    ),

    BatchNormalization(),

    Dropout(
        0.25
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )
])


ann_model.compile(

    optimizer=(
        tf.keras.optimizers.Adam(
            learning_rate=0.001
        )
    ),

    loss=(
        "sparse_categorical_crossentropy"
    ),

    metrics=[
        "accuracy"
    ]
)


ann_model.summary()


# ======================================================================
# 33. TRAIN ANN
# ======================================================================

early_stopping = (
    EarlyStopping(
        monitor="val_loss",
        patience=20,
        restore_best_weights=True,
        verbose=1
    )
)


print("\n" + "=" * 80)
print("TRAINING AIS-SELECTED ANN")
print("=" * 80)


history = ann_model.fit(

    X_train,
    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=4,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# ======================================================================
# 34. ANN PREDICTIONS
# ======================================================================

ann_probabilities = (
    ann_model.predict(
        X_test,
        verbose=0
    )
)


ann_predictions = (
    np.argmax(
        ann_probabilities,
        axis=1
    )
)


ann_accuracy = (
    accuracy_score(
        y_test,
        ann_predictions
    )
)


print(
    "\nAIS ANN Accuracy:",
    f"{ann_accuracy * 100:.2f}%"
)


# ======================================================================
# 35. SAVE AIS H5 MODEL
# ======================================================================

ann_model.save(
    H5_PATH
)


print(
    "\nSaved:",
    H5_PATH
)


# ======================================================================
# 36. SAVE AIS PKL
# ======================================================================

preprocessor = {

    "method":
        "Artificial Immune System Feature Selection",

    "algorithm":
        "Clonal Selection",

    "original_feature_columns":
        feature_columns,

    "selected_features":
        selected_features,

    "rejected_features":
        rejected_features,

    "best_antibody":
        best_antibody.tolist(),

    "best_fitness":
        float(best_fitness),

    "imputer":
        final_imputer,

    "scaler":
        final_scaler,

    "label_encoder":
        label_encoder,

    "class_names":
        class_names,

    "target_source_column":
        target_source_column,

    "state_column":
        state_column,

    "lower_threshold":
        float(lower_threshold),

    "upper_threshold":
        float(upper_threshold)
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        preprocessor,
        file
    )


print(
    "Saved:",
    PKL_PATH
)


# ======================================================================
# 37. AIS ACCURACY GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(
    history.history[
        "accuracy"
    ],
    linewidth=2,
    label="Training Accuracy"
)


plt.plot(
    history.history[
        "val_accuracy"
    ],
    linewidth=2,
    label="Validation Accuracy"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "AIS-ANN Training and Validation Accuracy",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 38. AIS HEATMAP
# ======================================================================

# Correlation heatmap only for AIS-selected variables + target source.

heatmap_columns = (
    selected_features
    +
    [
        target_source_column
    ]
)


# Remove duplicate names just in case.

heatmap_columns = list(
    dict.fromkeys(
        heatmap_columns
    )
)


correlation_matrix = (
    df[
        heatmap_columns
    ]
    .corr()
)


fig, ax = plt.subplots(
    figsize=(
        max(
            9,
            len(heatmap_columns)
            * 1.2
        ),
        max(
            7,
            len(heatmap_columns)
            * 1.0
        )
    )
)


image_plot = ax.imshow(
    correlation_matrix,
    aspect="auto"
)


fig.colorbar(
    image_plot,
    ax=ax,
    label="Correlation"
)


ax.set_xticks(
    np.arange(
        len(
            correlation_matrix.columns
        )
    )
)


ax.set_yticks(
    np.arange(
        len(
            correlation_matrix.columns
        )
    )
)


ax.set_xticklabels(
    correlation_matrix.columns,
    rotation=45,
    ha="right",
    fontsize=8
)


ax.set_yticklabels(
    correlation_matrix.columns,
    fontsize=8
)


for i in range(
    len(
        correlation_matrix.columns
    )
):

    for j in range(
        len(
            correlation_matrix.columns
        )
    ):

        value = (
            correlation_matrix
            .iloc[i, j]
        )


        if not np.isnan(value):

            ax.text(
                j,
                i,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=7
            )


ax.set_title(
    "AIS-Selected Feature Correlation Heatmap",
    fontsize=14,
    fontweight="bold",
    pad=15
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 39. CLASSICAL MACHINE LEARNING MODELS
# ======================================================================

print("\n" + "=" * 80)
print("AIS-SELECTED MODEL COMPARISON")
print("=" * 80)


models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000,
            random_state=RANDOM_STATE
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=RANDOM_STATE
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            random_state=RANDOM_STATE,
            class_weight="balanced"
        ),

    "KNN":
        KNeighborsClassifier(
            n_neighbors=3
        ),

    "SVM":
        SVC(
            kernel="rbf",
            probability=True,
            random_state=RANDOM_STATE
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=RANDOM_STATE
        )
}


model_names = []

model_accuracies = []

model_predictions = {}


for model_name, classifier in (
    models.items()
):

    print(
        "\nTraining:",
        model_name
    )


    classifier.fit(
        X_train,
        y_train
    )


    predictions = (
        classifier.predict(
            X_test
        )
    )


    accuracy = (
        accuracy_score(
            y_test,
            predictions
        )
    )


    model_names.append(
        model_name
    )

    model_accuracies.append(
        accuracy
    )

    model_predictions[
        model_name
    ] = predictions


    print(
        f"{model_name}: "
        f"{accuracy * 100:.2f}%"
    )


# Add ANN.

model_names.append(
    "Artificial Neural Network"
)

model_accuracies.append(
    ann_accuracy
)

model_predictions[
    "Artificial Neural Network"
] = (
    ann_predictions
)


# ======================================================================
# 40. AIS COMPARISON GRAPH
# ======================================================================

plt.figure(
    figsize=(12, 7)
)


accuracy_percentages = [

    value * 100

    for value in model_accuracies
]


bars = plt.bar(
    model_names,
    accuracy_percentages
)


plt.xlabel(
    "Classification Model"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.title(
    "AIS-Selected Feature Model Accuracy Comparison",
    fontsize=14,
    fontweight="bold"
)


plt.xticks(
    rotation=35,
    ha="right"
)


plt.ylim(
    0,
    110
)


for bar, value in zip(
    bars,
    accuracy_percentages
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center",
        va="bottom",
        fontsize=9
    )


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 41. AIS RESULT CSV
# ======================================================================

result_df = pd.DataFrame({

    "Model":
        model_names,

    "Accuracy":
        model_accuracies,

    "Accuracy_Percentage":
        [
            round(
                value * 100,
                2
            )

            for value
            in model_accuracies
        ],

    "AIS_Features_Selected":
        len(
            selected_features
        ),

    "Total_Candidate_Features":
        NUMBER_OF_FEATURES,

    "AIS_Best_Fitness":
        round(
            best_fitness,
            6
        )
})


result_df = (
    result_df
    .sort_values(
        by="Accuracy",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


result_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(result_df) + 1
    )
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print("\nAIS RESULT CSV:")

print(
    result_df
)


print(
    "\nSaved:",
    RESULT_CSV_PATH
)


# ======================================================================
# 42. AIS RESULT GRAPH
# ======================================================================

plt.figure(
    figsize=(12, 7)
)


bars = plt.barh(

    result_df[
        "Model"
    ],

    result_df[
        "Accuracy_Percentage"
    ]
)


plt.xlabel(
    "Accuracy (%)"
)

plt.ylabel(
    "Classification Model"
)

plt.title(
    "AIS-Based Classification Results",
    fontsize=14,
    fontweight="bold"
)


plt.xlim(
    0,
    110
)


for bar, value in zip(

    bars,

    result_df[
        "Accuracy_Percentage"
    ]
):

    plt.text(

        value + 1,

        bar.get_y()
        +
        bar.get_height() / 2,

        f"{value:.2f}%",

        va="center"
    )


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 43. AIS PREDICTION CSV
# ======================================================================

actual_labels = (
    label_encoder
    .inverse_transform(
        y_test
    )
)


predicted_labels = (
    label_encoder
    .inverse_transform(
        ann_predictions
    )
)


prediction_df = (
    pd.DataFrame()
)


if (
    state_column is not None
    and
    state_column in df.columns
):

    prediction_df[
        "State_Region"
    ] = (

        df.iloc[
            test_indices
        ][
            state_column
        ]
        .astype(str)
        .values
    )

else:

    prediction_df[
        "Sample"
    ] = [

        f"Sample_{i + 1}"

        for i in range(
            len(y_test)
        )
    ]


prediction_df[
    "Actual_Unmet_Need"
] = (

    df.iloc[
        test_indices
    ][
        target_source_column
    ]
    .values
)


prediction_df[
    "Actual_Category"
] = (
    actual_labels
)


prediction_df[
    "AIS_Predicted_Category"
] = (
    predicted_labels
)


prediction_df[
    "Correct_Prediction"
] = (

    prediction_df[
        "Actual_Category"
    ]
    ==
    prediction_df[
        "AIS_Predicted_Category"
    ]
)


prediction_df[
    "AIS_Prediction_Confidence"
] = (

    np.max(
        ann_probabilities,
        axis=1
    )
)


for class_index, class_name in enumerate(
    class_names
):

    prediction_df[
        f"AIS_Probability_{class_name}"
    ] = (

        ann_probabilities[
            :,
            class_index
        ]
    )


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print("\nAIS PREDICTIONS:")

print(
    prediction_df
)


print(
    "\nSaved:",
    PREDICTION_CSV_PATH
)


# ======================================================================
# 44. AIS PREDICTION GRAPH
# ======================================================================

if (
    "State_Region"
    in prediction_df.columns
):

    sample_labels = (

        prediction_df[
            "State_Region"
        ]
        .astype(str)
        .tolist()
    )

else:

    sample_labels = (

        prediction_df[
            "Sample"
        ]
        .astype(str)
        .tolist()
    )


x_positions = (
    np.arange(
        len(
            prediction_df
        )
    )
)


plt.figure(
    figsize=(
        max(
            12,
            len(
                prediction_df
            )
            * 1.2
        ),
        7
    )
)


plt.plot(
    x_positions,
    y_test,
    marker="o",
    linewidth=2,
    markersize=8,
    label="Actual Category"
)


plt.plot(
    x_positions,
    ann_predictions,
    marker="s",
    linestyle="--",
    linewidth=2,
    markersize=7,
    label="AIS-ANN Prediction"
)


plt.xticks(
    x_positions,
    sample_labels,
    rotation=45,
    ha="right"
)


plt.yticks(
    np.arange(
        len(
            class_names
        )
    ),
    class_names
)


plt.xlabel(
    "State / Test Sample"
)

plt.ylabel(
    "Unmet Need Category"
)

plt.title(
    "AIS-ANN Actual vs Predicted Unmet Need Category",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# ======================================================================
# 45. CONFUSION MATRIX
# ======================================================================

confusion = (
    confusion_matrix(
        y_test,
        ann_predictions,
        labels=np.arange(
            len(
                class_names
            )
        )
    )
)


print("\n" + "=" * 80)
print("AIS-ANN CONFUSION MATRIX")
print("=" * 80)

print(
    confusion
)


# ======================================================================
# 46. CLASSIFICATION REPORT
# ======================================================================

report_text = (
    classification_report(
        y_test,
        ann_predictions,
        labels=np.arange(
            len(
                class_names
            )
        ),
        target_names=class_names,
        zero_division=0
    )
)


print(
    "\nClassification Report:\n"
)

print(
    report_text
)


report_dictionary = (
    classification_report(
        y_test,
        ann_predictions,
        labels=np.arange(
            len(
                class_names
            )
        ),
        target_names=class_names,
        zero_division=0,
        output_dict=True
    )
)


# ======================================================================
# 47. BEST MODEL
# ======================================================================

best_model_index = int(
    np.argmax(
        model_accuracies
    )
)


best_model_name = (
    model_names[
        best_model_index
    ]
)


best_model_accuracy = (
    model_accuracies[
        best_model_index
    ]
)


print("\n" + "=" * 80)
print("BEST AIS-BASED MODEL")
print("=" * 80)


print(
    "\nModel:",
    best_model_name
)


print(
    "Accuracy:",
    f"{best_model_accuracy * 100:.2f}%"
)


# ======================================================================
# 48. SAVE YAML CONFIG
# ======================================================================

config = {

    "project": {

        "name":
            "State-Level Family Planning Unmet Need Pattern Analysis and Classification",

        "method":
            "Artificial Immune System",

        "dataset":
            DATASET_PATH,

        "output_directory":
            PROJECT_DIR
    },


    "ais": {

        "algorithm":
            "Clonal Selection",

        "population_size":
            POPULATION_SIZE,

        "generations":
            GENERATIONS,

        "elite_size":
            ELITE_SIZE,

        "clones_per_elite":
            CLONES_PER_ELITE,

        "mutation_rate":
            MUTATION_RATE,

        "random_new_antibodies":
            RANDOM_NEW_ANTIBODIES,

        "feature_penalty":
            FEATURE_PENALTY,

        "best_fitness":
            float(best_fitness),

        "total_candidate_features":
            int(
                NUMBER_OF_FEATURES
            ),

        "selected_feature_count":
            int(
                len(
                    selected_features
                )
            ),

        "selected_features":
            [
                str(x)
                for x
                in selected_features
            ],

        "rejected_features":
            [
                str(x)
                for x
                in rejected_features
            ]
    },


    "classification": {

        "target_source":
            str(
                target_source_column
            ),

        "target":
            "Unmet_Need_Category",

        "classes":
            [
                str(x)
                for x
                in class_names
            ],

        "lower_threshold":
            float(
                lower_threshold
            ),

        "upper_threshold":
            float(
                upper_threshold
            )
    },


    "ann": {

        "hidden_layers":
            [
                64,
                32,
                16
            ],

        "optimizer":
            "Adam",

        "loss":
            "Sparse Categorical Crossentropy",

        "epochs":
            200,

        "batch_size":
            4,

        "accuracy":
            float(
                ann_accuracy
            )
    },


    "best_model": {

        "name":
            str(
                best_model_name
            ),

        "accuracy":
            float(
                best_model_accuracy
            )
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nSaved:",
    YAML_PATH
)


# ======================================================================
# 49. SAVE JSON METADATA
# ======================================================================

metadata = {

    "project":
        "State-Level Family Planning Unmet Need Pattern Analysis and Classification",

    "optimization_method":
        "Artificial Immune System",

    "ais_algorithm":
        "Clonal Selection",

    "dataset_path":
        DATASET_PATH,

    "dataset_rows":
        int(
            len(df)
        ),

    "dataset_columns":
        int(
            len(
                df.columns
            )
        ),

    "original_columns":
        original_columns,

    "target_source_column":
        str(
            target_source_column
        ),

    "target_column":
        "Unmet_Need_Category",

    "class_names":
        [
            str(x)
            for x
            in class_names
        ],

    "candidate_features":
        [
            str(x)
            for x
            in feature_columns
        ],

    "selected_features":
        [
            str(x)
            for x
            in selected_features
        ],

    "rejected_features":
        [
            str(x)
            for x
            in rejected_features
        ],

    "best_antibody":
        [
            int(x)
            for x
            in best_antibody
        ],

    "best_ais_fitness":
        float(
            best_fitness
        ),

    "ann_accuracy":
        float(
            ann_accuracy
        ),

    "best_model":
        str(
            best_model_name
        ),

    "best_model_accuracy":
        float(
            best_model_accuracy
        ),

    "confusion_matrix":
        confusion.tolist(),

    "classification_report":
        report_dictionary
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "Saved:",
    JSON_PATH
)


# ======================================================================
# 50. VERIFY SAVED H5 MODEL
# ======================================================================

print("\n" + "=" * 80)
print("VERIFYING AIS H5 MODEL")
print("=" * 80)


loaded_model = (
    tf.keras.models.load_model(
        H5_PATH
    )
)


verification_probabilities = (
    loaded_model.predict(
        X_test[:1],
        verbose=0
    )
)


verification_prediction = int(
    np.argmax(
        verification_probabilities,
        axis=1
    )[0]
)


verification_label = (
    label_encoder
    .inverse_transform(
        [
            verification_prediction
        ]
    )[0]
)


print(
    "\nH5 model loaded successfully."
)

print(
    "Sample prediction:",
    verification_label
)


# ======================================================================
# 51. VERIFY PKL
# ======================================================================

print("\n" + "=" * 80)
print("VERIFYING AIS PKL")
print("=" * 80)


with open(
    PKL_PATH,
    "rb"
) as file:

    loaded_preprocessor = (
        pickle.load(
            file
        )
    )


print(
    "\nPKL loaded successfully."
)


print(
    "AIS selected features:"
)


for feature in (
    loaded_preprocessor[
        "selected_features"
    ]
):

    print(
        " -",
        feature
    )


# ======================================================================
# 52. VERIFY ALL OUTPUT FILES
# ======================================================================

output_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_CSV_PATH,

    FEATURE_SELECTION_GRAPH_PATH,

    FITNESS_GRAPH_PATH
]


print("\n" + "=" * 80)
print("VERIFYING ALL AIS OUTPUT FILES")
print("=" * 80)


for path in output_files:

    if os.path.exists(
        path
    ):

        size_kb = (
            os.path.getsize(
                path
            )
            /
            1024
        )


        print(
            f"\n[OK] "
            f"{os.path.basename(path)}"
        )

        print(
            f"     "
            f"{size_kb:.2f} KB"
        )

    else:

        print(
            f"\n[ERROR] "
            f"{os.path.basename(path)}"
        )


# ======================================================================
# 53. FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 80)
print("AIS PROJECT SUMMARY")
print("=" * 80)


print(
    f"""
PROJECT
-------
State-Level Family Planning Unmet Need Pattern Analysis and Classification


OPTIMIZATION METHOD
-------------------
Artificial Immune System (AIS)

AIS Algorithm:
Clonal Selection


DATASET
-------
{DATASET_PATH}


TARGET SOURCE
-------------
{target_source_column}


CLASSIFICATION
--------------
Low
Moderate
High


AIS FEATURE SELECTION
---------------------
Candidate Features:
{NUMBER_OF_FEATURES}

Selected Features:
{len(selected_features)}

Rejected Features:
{len(rejected_features)}

Best AIS Fitness:
{best_fitness:.4f}


SELECTED FEATURES
-----------------
"""
)


for feature in selected_features:

    print(
        " -",
        feature
    )


print(
    f"""

MODEL PERFORMANCE
-----------------
AIS-ANN Accuracy:
{ann_accuracy * 100:.2f}%

Best Model:
{best_model_name}

Best Accuracy:
{best_model_accuracy * 100:.2f}%


GENERATED FILES
---------------

MODEL FILES

1. ais_family_planning_model.h5

2. ais_family_planning_preprocessor.pkl

3. ais_family_planning_config.yaml

4. ais_family_planning_metadata.json


AIS FEATURE SELECTION FILES

5. ais_feature_selection.csv

6. ais_feature_selection_graph.png

7. ais_fitness_graph.png


MODEL VISUALIZATIONS

8. ais_accuracy_graph.png

9. ais_heatmap.png

10. ais_comparison_graph.png


RESULT FILES

11. ais_result.csv

12. ais_result_graph.png


PREDICTION FILES

13. ais_prediction.csv

14. ais_prediction_graph.png


OUTPUT DIRECTORY
----------------
{PROJECT_DIR}
"""
)


print("=" * 80)
print("AIS EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 80)
```
