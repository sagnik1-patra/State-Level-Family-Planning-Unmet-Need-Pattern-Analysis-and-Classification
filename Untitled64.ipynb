```python
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =====================================================================
# 1. PATHS
# =====================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\State-Level Family Planning Unmet Need Pattern Analysis and Classification"
)

DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "unmetneedforFPNFHS3.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)


# Output files

ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)

HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "comparison_graph.png"
)

RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)


# =====================================================================
# 2. LOAD DATASET
# =====================================================================

print("=" * 80)
print("STATE-LEVEL FAMILY PLANNING UNMET NEED ANALYSIS")
print("=" * 80)

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully.")
print("Original shape:", df.shape)

print("\nOriginal columns:")

for i, col in enumerate(df.columns, 1):
    print(i, col)


# =====================================================================
# 3. CLEAN COLUMN NAMES
# =====================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

df = df.dropna(
    axis=1,
    how="all"
)

print("\nCleaned shape:", df.shape)


# =====================================================================
# 4. DETECT STATE COLUMN
# =====================================================================

state_column = None

for col in df.columns:

    lower = col.lower()

    if (
        "state" in lower
        or "region" in lower
        or "area" in lower
    ):
        state_column = col
        break


# If no obvious state column found

if state_column is None:

    object_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    if len(object_columns) > 0:
        state_column = object_columns[0]


print("\nDetected State/Region column:")
print(state_column)


# =====================================================================
# 5. CONVERT POSSIBLE NUMERIC COLUMNS
# =====================================================================

for col in df.columns:

    if col == state_column:
        continue

    if df[col].dtype == "object":

        cleaned = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
            .str.replace("%", "", regex=False)
        )

        numeric_version = pd.to_numeric(
            cleaned,
            errors="coerce"
        )

        original_count = df[col].notna().sum()
        numeric_count = numeric_version.notna().sum()

        if (
            original_count > 0
            and numeric_count / original_count >= 0.60
        ):
            df[col] = numeric_version


# =====================================================================
# 6. NUMERIC COLUMNS
# =====================================================================

numeric_columns = df.select_dtypes(
    include=[np.number]
).columns.tolist()

print("\nNumeric columns:")

for col in numeric_columns:
    print(" -", col)

if len(numeric_columns) < 2:
    raise ValueError(
        "At least two numeric columns are required."
    )


# =====================================================================
# 7. DETECT TARGET SOURCE
# =====================================================================

target_source_column = None


# First preference:
# unmet + total

for col in numeric_columns:

    lower = col.lower()

    if (
        "unmet" in lower
        and "total" in lower
    ):
        target_source_column = col
        break


# Second preference:
# total

if target_source_column is None:

    for col in numeric_columns:

        if "total" in col.lower():
            target_source_column = col
            break


# Final fallback

if target_source_column is None:
    target_source_column = numeric_columns[0]


print("\nTarget source column:")
print(target_source_column)


# =====================================================================
# 8. REMOVE MISSING TARGET ROWS
# =====================================================================

df = df.dropna(
    subset=[target_source_column]
).reset_index(drop=True)


# =====================================================================
# 9. CREATE LOW / MODERATE / HIGH TARGET
# =====================================================================

target_values = df[
    target_source_column
].astype(float)


# Calculate thresholds explicitly so they can also be used later

lower_threshold = target_values.quantile(1 / 3)
upper_threshold = target_values.quantile(2 / 3)


def create_category(value):

    if value <= lower_threshold:
        return "Low"

    elif value <= upper_threshold:
        return "Moderate"

    else:
        return "High"


df["Unmet_Need_Category"] = (
    target_values.apply(
        create_category
    )
)


print("\nCategory thresholds:")

print(
    f"Low      : <= {lower_threshold:.3f}"
)

print(
    f"Moderate : > {lower_threshold:.3f} "
    f"and <= {upper_threshold:.3f}"
)

print(
    f"High     : > {upper_threshold:.3f}"
)


print("\nCategory distribution:")

print(
    df[
        "Unmet_Need_Category"
    ].value_counts()
)


# =====================================================================
# 10. LABEL ENCODING
# =====================================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df[
        "Unmet_Need_Category"
    ].astype(str)
)

class_names = (
    label_encoder
    .classes_
    .tolist()
)

print("\nClass mapping:")

for i, name in enumerate(class_names):
    print(i, "=", name)


# =====================================================================
# 11. FEATURES
# =====================================================================

# Remove target source to prevent target leakage.

feature_columns = [
    col
    for col in numeric_columns
    if col != target_source_column
]


if len(feature_columns) == 0:
    raise ValueError(
        "No predictor columns remain."
    )


X = df[
    feature_columns
].copy()

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


print("\nFeatures used:")

for feature in feature_columns:
    print(" -", feature)


# =====================================================================
# 12. IMPUTATION
# =====================================================================

imputer = SimpleImputer(
    strategy="median"
)

X_imputed = imputer.fit_transform(
    X
)


# =====================================================================
# 13. STANDARDIZATION
# =====================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X_imputed
)


# =====================================================================
# 14. TRAIN TEST SPLIT
# =====================================================================

unique_classes, class_counts = np.unique(
    y,
    return_counts=True
)

if class_counts.min() >= 2:
    stratify_value = y
else:
    stratify_value = None


X_train, X_test, y_train, y_test, train_idx, test_idx = (
    train_test_split(
        X_scaled,
        y,
        np.arange(len(df)),
        test_size=0.25,
        random_state=42,
        stratify=stratify_value
    )
)


print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# =====================================================================
# 15. BUILD ANN
# =====================================================================

number_of_features = X_train.shape[1]
number_of_classes = len(class_names)


model = tf.keras.Sequential([

    tf.keras.Input(
        shape=(number_of_features,)
    ),

    tf.keras.layers.Dense(
        64,
        activation="relu"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.Dropout(
        0.25
    ),

    tf.keras.layers.Dense(
        32,
        activation="relu"
    ),

    tf.keras.layers.Dropout(
        0.20
    ),

    tf.keras.layers.Dense(
        16,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        number_of_classes,
        activation="softmax"
    )
])


model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


# =====================================================================
# 16. TRAIN ANN
# =====================================================================

early_stopping = tf.keras.callbacks.EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True
)


print("\n" + "=" * 80)
print("TRAINING ARTIFICIAL NEURAL NETWORK")
print("=" * 80)


history = model.fit(

    X_train,
    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=4,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# =====================================================================
# 17. ANN PREDICTIONS
# =====================================================================

ann_probabilities = model.predict(
    X_test,
    verbose=0
)

ann_predictions = np.argmax(
    ann_probabilities,
    axis=1
)

ann_accuracy = accuracy_score(
    y_test,
    ann_predictions
)


print(
    f"\nANN Accuracy: "
    f"{ann_accuracy * 100:.2f}%"
)


# =====================================================================
# 18. ACCURACY GRAPH
# =====================================================================

print("\nGenerating accuracy graph...")


plt.figure(
    figsize=(10, 6)
)

plt.plot(
    history.history["accuracy"],
    linewidth=2,
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    linewidth=2,
    label="Validation Accuracy"
)

plt.xlabel(
    "Epoch",
    fontsize=12
)

plt.ylabel(
    "Accuracy",
    fontsize=12
)

plt.title(
    "ANN Training and Validation Accuracy",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


print(
    "Saved:",
    ACCURACY_GRAPH_PATH
)


# =====================================================================
# 19. CORRELATION HEATMAP
# =====================================================================
#
# Created using matplotlib only, so seaborn is not required.
# =====================================================================

print("\nGenerating correlation heatmap...")


correlation_data = df[
    numeric_columns
].copy()

correlation_matrix = (
    correlation_data.corr()
)


fig, ax = plt.subplots(
    figsize=(12, 9)
)

heatmap = ax.imshow(
    correlation_matrix,
    aspect="auto"
)

fig.colorbar(
    heatmap,
    ax=ax,
    label="Correlation"
)


ax.set_xticks(
    np.arange(
        len(correlation_matrix.columns)
    )
)

ax.set_yticks(
    np.arange(
        len(correlation_matrix.columns)
    )
)


ax.set_xticklabels(
    correlation_matrix.columns,
    rotation=45,
    ha="right",
    fontsize=8
)

ax.set_yticklabels(
    correlation_matrix.columns,
    fontsize=8
)


# Add values to heatmap

for i in range(
    len(correlation_matrix.columns)
):

    for j in range(
        len(correlation_matrix.columns)
    ):

        value = correlation_matrix.iloc[i, j]

        if not np.isnan(value):

            ax.text(
                j,
                i,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=7
            )


ax.set_title(
    "Correlation Heatmap of Family Planning Indicators",
    fontsize=14,
    fontweight="bold",
    pad=15
)

plt.tight_layout()

plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


print(
    "Saved:",
    HEATMAP_PATH
)


# =====================================================================
# 20. MODEL COMPARISON
# =====================================================================

print("\n" + "=" * 80)
print("TRAINING COMPARISON MODELS")
print("=" * 80)


models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000,
            random_state=42
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            random_state=42
        ),

    "KNN":
        KNeighborsClassifier(
            n_neighbors=3
        ),

    "SVM":
        SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=42
        )
}


model_names = []
model_accuracies = []


for model_name, classifier in models.items():

    print(
        "\nTraining:",
        model_name
    )

    classifier.fit(
        X_train,
        y_train
    )

    predictions = classifier.predict(
        X_test
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    model_names.append(
        model_name
    )

    model_accuracies.append(
        accuracy
    )

    print(
        f"{model_name} Accuracy: "
        f"{accuracy * 100:.2f}%"
    )


# Add ANN

model_names.append(
    "Artificial Neural Network"
)

model_accuracies.append(
    ann_accuracy
)


# =====================================================================
# 21. COMPARISON GRAPH
# =====================================================================

print("\nGenerating comparison graph...")


plt.figure(
    figsize=(12, 7)
)

bars = plt.bar(
    model_names,
    [
        x * 100
        for x in model_accuracies
    ]
)


plt.xlabel(
    "Machine Learning Model",
    fontsize=12
)

plt.ylabel(
    "Accuracy (%)",
    fontsize=12
)

plt.title(
    "Model Accuracy Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(
    rotation=35,
    ha="right"
)

plt.ylim(
    0,
    110
)


# Accuracy labels

for bar, accuracy in zip(
    bars,
    model_accuracies
):

    height = bar.get_height()

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        height + 1,

        f"{accuracy * 100:.2f}%",

        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


# =====================================================================
# 22. CREATE RESULT.CSV
# =====================================================================

print("\nCreating result.csv...")


result_df = pd.DataFrame({

    "Model":
        model_names,

    "Accuracy":
        model_accuracies,

    "Accuracy_Percentage":
        [
            round(x * 100, 2)
            for x in model_accuracies
        ]
})


result_df = result_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(
    drop=True
)


result_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(result_df) + 1
    )
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print("\nRESULT CSV:")
print(result_df)

print(
    "\nSaved:",
    RESULT_CSV_PATH
)


# =====================================================================
# 23. RESULT GRAPH
# =====================================================================

print("\nGenerating result graph...")


plt.figure(
    figsize=(12, 7)
)

bars = plt.barh(

    result_df["Model"],

    result_df[
        "Accuracy_Percentage"
    ]
)


plt.xlabel(
    "Accuracy (%)",
    fontsize=12
)

plt.ylabel(
    "Model",
    fontsize=12
)

plt.title(
    "Overall Classification Results",
    fontsize=15,
    fontweight="bold"
)

plt.xlim(
    0,
    110
)


for bar, value in zip(

    bars,

    result_df[
        "Accuracy_Percentage"
    ]
):

    plt.text(

        value + 1,

        bar.get_y()
        + bar.get_height() / 2,

        f"{value:.2f}%",

        va="center",
        fontsize=10
    )


plt.grid(
    axis="x",
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


# =====================================================================
# 24. CREATE PREDICTION.CSV
# =====================================================================

print("\nCreating prediction.csv...")


actual_labels = (
    label_encoder.inverse_transform(
        y_test
    )
)

predicted_labels = (
    label_encoder.inverse_transform(
        ann_predictions
    )
)


prediction_df = pd.DataFrame()


# Add State/Region if available

if (
    state_column is not None
    and state_column in df.columns
):

    prediction_df[
        "State_Region"
    ] = (
        df.iloc[test_idx][
            state_column
        ]
        .astype(str)
        .values
    )

else:

    prediction_df[
        "Sample"
    ] = [
        f"Sample_{i + 1}"
        for i in range(
            len(y_test)
        )
    ]


prediction_df[
    "Actual_Category"
] = actual_labels


prediction_df[
    "Predicted_Category"
] = predicted_labels


prediction_df[
    "Correct_Prediction"
] = (
    prediction_df[
        "Actual_Category"
    ]
    ==
    prediction_df[
        "Predicted_Category"
    ]
)


# Add original target values

prediction_df[
    "Actual_Unmet_Need_Value"
] = (
    df.iloc[test_idx][
        target_source_column
    ]
    .values
)


# Add probability for every class

for class_index, class_name in enumerate(
    class_names
):

    prediction_df[
        f"Probability_{class_name}"
    ] = ann_probabilities[
        :,
        class_index
    ]


# Maximum confidence

prediction_df[
    "Prediction_Confidence"
] = np.max(
    ann_probabilities,
    axis=1
)


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print("\nPREDICTION CSV:")
print(prediction_df)

print(
    "\nSaved:",
    PREDICTION_CSV_PATH
)


# =====================================================================
# 25. PREDICTION GRAPH
# =====================================================================
#
# Compare actual and predicted category numerically.
#
# Low / Moderate / High are converted to their encoded values.
# =====================================================================

print("\nGenerating prediction graph...")


sample_labels = None

if "State_Region" in prediction_df.columns:

    sample_labels = (
        prediction_df[
            "State_Region"
        ]
        .astype(str)
        .tolist()
    )

else:

    sample_labels = (
        prediction_df[
            "Sample"
        ]
        .astype(str)
        .tolist()
    )


x_positions = np.arange(
    len(prediction_df)
)


plt.figure(
    figsize=(
        max(
            12,
            len(prediction_df) * 1.2
        ),
        7
    )
)


plt.plot(

    x_positions,

    y_test,

    marker="o",

    linewidth=2,

    markersize=8,

    label="Actual Category"
)


plt.plot(

    x_positions,

    ann_predictions,

    marker="s",

    linestyle="--",

    linewidth=2,

    markersize=7,

    label="Predicted Category"
)


plt.xticks(

    x_positions,

    sample_labels,

    rotation=45,

    ha="right"
)


plt.yticks(

    np.arange(
        len(class_names)
    ),

    class_names
)


plt.xlabel(
    "State / Test Sample",
    fontsize=12
)

plt.ylabel(
    "Unmet Need Category",
    fontsize=12
)

plt.title(
    "Actual vs Predicted Family Planning Unmet Need Category",
    fontsize=14,
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()

plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


# =====================================================================
# 26. CONFUSION MATRIX
# =====================================================================

print("\n" + "=" * 80)
print("ANN CONFUSION MATRIX")
print("=" * 80)


cm = confusion_matrix(
    y_test,
    ann_predictions
)

print(cm)


# =====================================================================
# 27. CLASSIFICATION REPORT
# =====================================================================

print("\nClassification Report:\n")


print(

    classification_report(

        y_test,

        ann_predictions,

        labels=np.arange(
            len(class_names)
        ),

        target_names=class_names,

        zero_division=0
    )
)


# =====================================================================
# 28. PRINT BEST MODEL
# =====================================================================

best_index = int(
    np.argmax(
        model_accuracies
    )
)

best_model = model_names[
    best_index
]

best_accuracy = model_accuracies[
    best_index
]


print("\n" + "=" * 80)
print("BEST MODEL")
print("=" * 80)

print(
    "\nModel:",
    best_model
)

print(
    "Accuracy:",
    f"{best_accuracy * 100:.2f}%"
)


# =====================================================================
# 29. VERIFY ALL OUTPUT FILES
# =====================================================================

print("\n" + "=" * 80)
print("VERIFYING GENERATED FILES")
print("=" * 80)


output_files = [

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH
]


for file_path in output_files:

    if os.path.exists(file_path):

        size_kb = (
            os.path.getsize(
                file_path
            )
            / 1024
        )

        print(
            f"\n[OK] "
            f"{os.path.basename(file_path)}"
        )

        print(
            f"     Size: "
            f"{size_kb:.2f} KB"
        )

    else:

        print(
            f"\n[ERROR] "
            f"{os.path.basename(file_path)}"
        )


# =====================================================================
# 30. FINAL SUMMARY
# =====================================================================

print("\n" + "=" * 80)
print("PROJECT OUTPUT SUMMARY")
print("=" * 80)


print(
    f"""
Project:
State-Level Family Planning Unmet Need Pattern Analysis and Classification

Dataset:
{DATASET_PATH}

Target:
{target_source_column}

Classification:
Low / Moderate / High

Number of features:
{len(feature_columns)}

Training samples:
{len(X_train)}

Testing samples:
{len(X_test)}

ANN Accuracy:
{ann_accuracy * 100:.2f}%

Best Model:
{best_model}

Best Accuracy:
{best_accuracy * 100:.2f}%


GENERATED FILES
---------------

1. accuracy_graph.png
   ANN training vs validation accuracy

2. heatmap.png
   Correlation heatmap of numeric family planning indicators

3. comparison_graph.png
   Accuracy comparison of all ML/DL models

4. result.csv
   Model-wise accuracy and ranking

5. result_graph.png
   Overall model result visualization

6. prediction.csv
   Actual vs predicted category,
   prediction correctness,
   confidence and class probabilities

7. prediction_graph.png
   Actual vs predicted unmet-need categories


OUTPUT DIRECTORY
----------------

{PROJECT_DIR}

"""
)


print("=" * 80)
print("ALL FILES CREATED SUCCESSFULLY")
print("=" * 80)
```
